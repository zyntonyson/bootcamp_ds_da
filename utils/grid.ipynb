```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator, FuncFormatter

# --- Parámetros de salida ---
width_inch, height_inch = 8, 6   # tamaño físico de la imagen
dpi = 300                        # resolución
outfile = "grid_6x4_ticks_on_axes.jpeg"

# --- Conversión a milímetros ---
inch_to_mm = 25.4
width_mm  = width_inch  * inch_to_mm
height_mm = height_inch * inch_to_mm

# --- Figura y ejes ---
fig, ax = plt.subplots(figsize=(width_inch, height_inch), dpi=dpi)

# Límites centrados en (0,0)
ax.set_xlim(-width_mm/2,  width_mm/2)
ax.set_ylim(-height_mm/2, height_mm/2)

# --- Ticks y grid ---
major_step_mm = 10   # cada 10 mm con etiquetas
minor_step_mm = 2    # cada 2 mm para grid fino

ax.xaxis.set_major_locator(MultipleLocator(base=major_step_mm))
ax.yaxis.set_major_locator(MultipleLocator(base=major_step_mm))
ax.xaxis.set_minor_locator(MultipleLocator(base=minor_step_mm))
ax.yaxis.set_minor_locator(MultipleLocator(base=minor_step_mm))

fmt_mm = FuncFormatter(lambda v, pos: f"{int(np.round(v))}")
ax.xaxis.set_major_formatter(fmt_mm)
ax.yaxis.set_major_formatter(fmt_mm)

# Grid fino (2 mm) + grueso (10 mm)
ax.grid(which="minor", linewidth=0.3)
ax.grid(which="major", linewidth=0.8)

# --- Ejes al centro ---
ax.spines["left"].set_position("center")
ax.spines["bottom"].set_position("center")
ax.spines["right"].set_color("none")
ax.spines["top"].set_color("none")

# Ticks sobre los ejes (no en bordes)
ax.xaxis.set_ticks_position("bottom")
ax.yaxis.set_ticks_position("left")

# Etiquetas
#ax.set_xlabel("Eje X (mm)", loc="right")
#ax.set_ylabel("Eje Y (mm)", loc="top")

# Origen
ax.plot(0, 0, "o", markersize=3, color="black")
ax.text(2, 2, "(0,0)", fontsize=8)

# Aspecto cuadrado
ax.set_aspect("equal", adjustable="box")

# Guardar como JPEG
plt.savefig(outfile, format="jpeg", dpi=dpi, bbox_inches="tight")
plt.close()

print(f"Imagen guardada en: {outfile}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator, FuncFormatter

# --- Parámetros físicos ---
width_inch, height_inch = 6, 4   # tamaño físico
dpi = 300                        # resolución de salida
outfile = "grid_real_mm.jpeg"

# Conversión a mm
inch_to_mm = 25.4
width_mm  = width_inch  * inch_to_mm
height_mm = height_inch * inch_to_mm

# --- Crear figura ---
fig, ax = plt.subplots(figsize=(width_inch, height_inch), dpi=dpi)

# Límites centrados en (0,0) expresados en mm reales
ax.set_xlim(-width_mm/2,  width_mm/2)
ax.set_ylim(-height_mm/2, height_mm/2)

# --- Ticks ---
major_mm = 10   # etiquetas cada 10 mm
minor_mm = 5    # grid cada 5 mm

ax.xaxis.set_major_locator(MultipleLocator(major_mm))
ax.yaxis.set_major_locator(MultipleLocator(major_mm))
ax.xaxis.set_minor_locator(MultipleLocator(minor_mm))
ax.yaxis.set_minor_locator(MultipleLocator(minor_mm))

# Formatear etiquetas en mm
fmt = FuncFormatter(lambda v, pos: f"{int(v)}")
ax.xaxis.set_major_formatter(fmt)
ax.yaxis.set_major_formatter(fmt)

# --- Grid ---
ax.grid(which="minor", linewidth=0.3, color="gray")
ax.grid(which="major", linewidth=0.8, color="black")

# --- Ejes al centro ---
ax.spines["left"].set_position("center")
ax.spines["bottom"].set_position("center")
ax.spines["right"].set_color("none")
ax.spines["top"].set_color("none")

# Ticks sobre los ejes, no en bordes
ax.xaxis.set_ticks_position("bottom")
ax.yaxis.set_ticks_position("left")

# Etiquetas
ax.set_xlabel( "",loc="right",fontsize=6)
ax.set_ylabel( "",loc="top",fontsize=6)

# Marcar origen
ax.plot(0, 0, "o", color="black", markersize=3)
ax.text(2, 2, "(0,0)", fontsize=8)

# Relación de aspecto
ax.set_aspect("equal", adjustable="box")


# Eliminar márgenes internos
ax.margins(0)

# Guardar imagen
plt.savefig(outfile, format="jpeg", dpi=dpi, bbox_inches="tight", pad_inches=0)
plt.close()

print(f"Imagen guardada en: {outfile}")
```
