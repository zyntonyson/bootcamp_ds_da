````python
from sqlalchemy import create_engine, inspect
import pandas as pd

# ========================
# 1️⃣ Configuración de PostgreSQL
# ========================
db_config = {
    'user': 'practicum_student',
    'pwd': 'QnmDH8Sc2TQLvy2G3Vvh7',
    'host': 'yp-trainers-practicum.cluster-czs0gxyx2d8w.us-east-1.rds.amazonaws.com',
    'port': 5432,
    'db': 'data-analyst-final-project-db'
}

connection_string = f"postgresql://{db_config['user']}:{db_config['pwd']}@{db_config['host']}:{db_config['port']}/{db_config['db']}"
engine = create_engine(connection_string, connect_args={'sslmode': 'require'})

# ========================
# 2️⃣ Inspeccionar las tablas
# ========================
inspector = inspect(engine)
tables = inspector.get_table_names()
print(f"📋 Tablas detectadas: {tables}")

# ========================
# 3️⃣ Generar estructura Mermaid
# ========================
mermaid = ["erDiagram"]

for table in tables:
    mermaid.append(f"    {table} {{")
    columns = inspector.get_columns(table)
    pks = inspector.get_pk_constraint(table)['constrained_columns']

    for col in columns:
        colname = col['name']
        dtype = str(col['type'])
        if colname in pks:
            colname = f"*{colname}"  # marca clave primaria
        mermaid.append(f"        {dtype} {colname}")
    mermaid.append("    }")

# ========================
# 4️⃣ Agregar relaciones (foreign keys)
# ========================
for table in tables:
    fks = inspector.get_foreign_keys(table)
    for fk in fks:
        ref_table = fk['referred_table']
        local_cols = ", ".join(fk['constrained_columns'])
        ref_cols = ", ".join(fk['referred_columns'])
        mermaid.append(f"    {ref_table} ||--o{{ {table} : {ref_cols} → {local_cols}")

# ========================
# 5️⃣ Guardar el resultado
# ========================
mermaid_diagram = "\n".join(mermaid)
output_file = "database_erd_mermaid.md"

with open(output_file, "w", encoding="utf-8") as f:
    f.write("```mermaid\n")
    f.write(mermaid_diagram)
    f.write("\n```")

print("\n✅ Diagrama ER generado en formato Mermaid.")
print(f"📄 Archivo: {output_file}")
print("\n🔍 Puedes visualizarlo en:")
print(" - VSCode (extensión Markdown Preview Mermaid Support)")
print(" - GitHub (abre el .md directamente)")
print(" - Notion o Mermaid Live Editor: https://mermaid.live")
````
