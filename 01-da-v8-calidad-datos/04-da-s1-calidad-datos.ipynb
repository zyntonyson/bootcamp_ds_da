# 🚗 Webinar – Calidad de Datos: Accidentes Viales

## 🎯 Objetivo de la sesión

Trabajar en equipos para **detectar, documentar y limpiar problemas de calidad de datos** en un dataset realista de accidentes viales.


## 🧠 Análisis de accidentes viales

Una dependencia de movilidad urbana quiere analizar factores asociados a la gravedad de accidentes.

Sin embargo, el dataset recibido presenta:

- Fechas en múltiples formatos
- Horas inválidas
- Categorías mal escritas
- Valores faltantes
- Booleanos inconsistentes
- Números mezclados con texto
- Posibles errores geográficos

Antes de analizar… necesitamos limpiar.



### 🧹 Parte 1 – Limpieza de datos (Trabajo en equipos)

 *👥 Organización*

- Equipos de 3–4 personas
- Cada equipo trabaja sobre la misma base
- Tiempo estimado: 60–90 minutos
- Datos: [Accidentes viales](https://docs.google.com/spreadsheets/d/1l3bScY898FOxZki9KV6t_kLKyg3EclyIFWijQL47vDM/copy)



#### 🔍 Fase 1: Diagnóstico inicial

Cada equipo debe responder:

- ¿Cuántos registros tiene el dataset?
- ¿Existen registros duplicados?
- ¿Qué columnas tienen valores faltantes?
- ¿Qué columnas tienen inconsistencias visibles?
- ¿Qué columnas requieren conversión de tipo?

> 💡 Consejo: no empiecen limpiando. Primero entiendan el problema.



#### 🗂️ Fase 2: Estandarización de formatos

##### 📅 Fecha y hora

- Unificar `fecha_accidente` a formato YYYY-MM-DD
- Detectar fechas inválidas
- Convertir horas a formato 24h válido
- Detectar horas imposibles



##### 🔤 Categóricas

Estandarizar:

- ciudad
- tipo_via
- tipo_accidente
- clima
- gravedad
- genero
- uso_cinturon
- iluminacion
- trafico

Ejemplo de regla:

- "Colision", "colisión", "Colisión" → **colisión**
- "Sí", "si", "SÍ" → **Sí**



##### 🔢 Numéricas

Limpiar:

- limite_velocidad (remover texto como “km/h”)
- visibilidad_metros (remover “m”)
- prueba_alcohol (cambiar coma por punto)
- edad_conductor
- anos_experiencia

Revisar:

- Valores fuera de rango
- Negativos
- Texto en campos numéricos



##### ☑️ Booleanos

Unificar:

- zona_escolar
- licencia_vigente

Ejemplo:

TRUE, true, 1, Sí → TRUE  
FALSE, false, 0, No → FALSE  



### 📊 Parte 2 – Validación de limpieza

Cada equipo debe reportar:

- Cantidad final de registros
- Cantidad de valores faltantes restantes
- Número de categorías únicas por variable clave
- Transformaciones aplicadas



### 🔎 Parte 3 – Mini análisis exploratorio

Con los datos limpios responder:

- ¿Qué ciudad tiene más accidentes graves?
- ¿Qué tipo de vía concentra mayor proporción de accidentes fatales?
- ¿Existe relación aparente entre alcohol y gravedad?
- ¿Hay más accidentes graves en horario nocturno?
- ¿Qué clima se asocia más a accidentes graves?
