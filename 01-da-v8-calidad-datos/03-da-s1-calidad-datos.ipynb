Eres el nuevo *Data Analyst* de esta startup mexicana que busca revolucionar el mercado de rentas vacacionales. **Ana**, la Directora de Expansión, te ha convocado a una reunión urgente. 🚨


MexStay está planeando su expansión agresiva para la próxima temporada alta. Para tomar decisiones inteligentes, el equipo de ingeniería hizo un *web scraping* masivo de las plataformas de competencia (como Airbnb y Vrbo) para entender el mercado.

Sin embargo, hay un **gran problema**: La información se extrajo de fuentes muy diversas y el dataset resultante está **"sucio"**. 😫

Ana te comenta:
> *"Necesitamos saber urgentemente cuál es el precio promedio de alojamiento en ciudades de interés de la República y qué propiedades tienen mejor rating, pero el Excel que tenemos es inusable. Hay precios que parecen fechas, fechas que parecen texto y los servicios están todos mezclados. ¡Ayúdanos a poner orden para poder decidir dónde invertir!"*

### 🚀 Tu Misión
Tu objetivo es transformar este dataset crudo y caótico en una fuente de verdad confiable. Sin una limpieza adecuada, cualquier análisis de precios o popularidad será erróneo y podría costar millones a la empresa.

### 📂 Sobre los Datos
El dataset contiene una muestra de **50 alojamientos** extraídos automáticamente para realizar las pruebas de limpieza que serán aplicadas posteriormente de forma masiva.
*   **Origen:** Web scraping de múltiples sitios de reserva y archivos CSV legacy.
*   **Estructura:**
    *   `precio_noche`: Crítico para el análisis financiero, pero viene con formatos de moneda mixtos (estilo europeo vs americano).
    *   `rating`: Fundamental para filtrar calidad, pero inconsistente en el uso de decimales (puntos y comas).
    *   `servicios`: Una lista de texto separada por comas que necesita estandarización para saber quién ofrece realmente "Aire Acondicionado".
    *   `ultima_reserva`: Fechas en formatos variados (Español, Inglés, ISO) que impiden calcular la recencia.

## 🛠️ Guía de Operaciones: Protocolo de Limpieza de Datos

Sigue estos pasos para transformar los datos crudos en información lista para el análisis. Recuerda: **La calidad del análisis depende de la calidad de los datos.**

### 1️⃣ Preparación del Entorno 🛡️
*   **Preserva la fuente:** Haz una copia de la hoja `raw` y renómbrala como `clean_data`.
    *   *Nota:* Todas las transformaciones se harán exclusivamente en `clean_data`.

### 2️⃣ Estructuración y Limpieza (Hoja `clean_data`) 🧹
Una vez en tu nueva hoja de trabajo, ejecuta las siguientes tareas:

*   **A. Separación de Datos:**
    *   La información viene comprimida. Utiliza la herramienta de texto en columnas para separar los campos usando el delimitador correcto (punto y coma `;`).

*   **B. Estandarización de Texto:**
    *   **Espacios:** Elimina los espacios innecesarios al inicio y al final de las cadenas (Trim).
    *   **Formato:** Convierte el texto a un formato uniforme (sugerencia: *Minúsculas* o *Nombre Propio*) para evitar duplicados por mayúsculas (ej. "Tulum" vs "TULUM").

*   **C. Normalización Numérica y Fechas:** 🔢
    *   **Precios y Ratings:** Asegúrate de que Google Sheets los reconozca como números y no como texto. Unifica los separadores decimales (punto vs coma).
    *   **Fechas:** Estandariza todas las fechas a un formato único (ej. `AAAA-MM-DD`) para poder ordenar cronológicamente.

*   **D. Enriquecimiento Geográfico:** 📍
    *   Separa la columna `ciudad_estado` en dos columnas independientes: `ciudad` y `estado`.

*   **E. Ingeniería de Características (Feature Engineering):** ⚡
    *   Analiza la columna de servicios y crea 3 nuevas columnas booleanas (Verdadero/Falso) o binarias (1/0):
        *   `tiene_wifi` 📶
        *   `tiene_ac` ❄️
        *   `tiene_piscina` 🏊
    
    > 💡 **Tip:** Si tienes dudas sobre qué fórmula usar para buscar texto dentro de una celda (como `REGEXMATCH` o `HALLAR`), ¡pregúntale a la IA!

### 3️⃣ Reporte de Resultados 📊
*   Ve a la hoja `resumen`.
*   Utiliza tus datos limpios para calcular las métricas y llenar la información solicitada por el equipo de negocio.
