Eres el nuevo *Data Analyst* de esta startup mexicana que busca revolucionar el mercado de rentas vacacionales. **Ana**, la Directora de Expansión, te ha convocado a una reunión urgente. 🚨


MexStay está planeando su expansión agresiva para la próxima temporada alta. Para tomar decisiones inteligentes, el equipo de ingeniería hizo un *web scraping* masivo de las plataformas de competencia (como Airbnb y Vrbo) para entender el mercado.

Sin embargo, hay un **gran problema**: La información se extrajo de fuentes muy diversas y el dataset resultante está **"sucio"**. 😫

Ana te comenta:
> *"Necesitamos saber urgentemente cuál es el precio promedio de alojamiento en ciudades de interés de la República y qué propiedades tienen mejor rating, pero el Excel que tenemos es inusable. Hay precios que parecen fechas, fechas que parecen texto y los servicios están todos mezclados. ¡Ayúdanos a poner orden para poder decidir dónde invertir!"*

### 🚀 Tu Misión
Tu objetivo es transformar este dataset crudo y caótico en una fuente de verdad confiable. Sin una limpieza adecuada, cualquier análisis de precios o popularidad será erróneo y podría costar millones a la empresa.

### 📂 Sobre los Datos
El dataset contiene una muestra de **50 alojamientos** extraídos automáticamente para realizar las pruenas de limpieza que serán aplicadas posteriormente de forma masiva.
*   **Origen:** Web scraping de múltiples sitios de reserva y archivos CSV legacy.
*   **Estructura:**
    *   `precio_noche`: Crítico para el análisis financiero, pero viene con formatos de moneda mixtos (estilo europeo vs americano).
    *   `rating`: Fundamental para filtrar calidad, pero inconsistente en el uso de decimales (puntos y comas).
    *   `servicios`: Una lista de texto separada por comas que necesita estandarización para saber quién ofrece realmente "Aire Acondicionado".
    *   `ultima_reserva`: Fechas en formatos variados (Español, Inglés, ISO) que impiden calcular la recencia.
