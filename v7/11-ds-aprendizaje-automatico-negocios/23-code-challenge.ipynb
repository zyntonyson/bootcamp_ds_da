### 📁 **Descripción del dataset**

Cada fila representa una **sesión individual** de un visitante en el sitio. Las columnas incluyen:

- Páginas visitadas por tipo (`Administrative`, `ProductRelated`, etc.)
- Tiempos de duración por tipo de página
- Tasas de rebote y salida (`BounceRates`, `ExitRates`)
- `PageValues`: estimación monetaria de la sesión
- Datos técnicos (`OperatingSystems`, `TrafficType`, `Region`, etc.)
- `Revenue`: indica si esa sesión terminó en compra (`TRUE`/`FALSE`)
