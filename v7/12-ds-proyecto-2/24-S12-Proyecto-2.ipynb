## 🗂️ Diccionario de datos Salaries[^1]

---



| Columna              | Descripción                                                                 |
|----------------------|-----------------------------------------------------------------------------|
| `work_year`          | Año en que se registró el dato (e.g., 2022).                                |
| `experience_level`   | Nivel de experiencia del empleado (`EN`, `MI`, `SE`, `EX`).                 |
| `employment_type`    | Tipo de empleo (`FT`, `PT`, `CT`, `FL`).                                    |
| `job_title`          | Título del puesto del empleado (e.g., `Data Scientist`, `ML Engineer`).     |
| `salary`             | Salario reportado en la moneda local.                                       |
| `salary_currency`    | Moneda del salario (e.g., `USD`, `EUR`, `INR`).                             |
| `salary_in_usd`      | Salario convertido a dólares estadounidenses (USD).                         |
| `employee_residence` | País de residencia del empleado (ISO country code, e.g., `US`, `DE`).       |
| `remote_ratio`       | Porcentaje de tiempo trabajado de forma remota (`0`, `50`, `100`).          |
| `company_location`   | País donde se ubica la empresa (ISO country code, e.g., `US`, `GB`).        |
| `company_size`       | Tamaño de la empresa (`S` = Pequeña, `M` = Mediana, `L` = Grande).          |

> 📝 Nota: No hay valores faltantes en el dataset. Todas las columnas están completas.





[^1]: Datos obtenidos de https://www.kaggle.com/datasets/rohit265/prediction-of-salary
