```python
import matplotlib.pyplot as plt
import numpy as np
import gradio as gr
import tensorflow as tf
from PIL import Image
import seaborn as sns
from sklearn.datasets import load_sample_image
from sklearn.metrics import confusion_matrix
from tensorflow.keras import layers, models
```

```python

# Cargamos una imagen de ejemplo ("flower.jpg")
flower = load_sample_image("flower.jpg")

# Mostramos la imagen original
plt.imshow(flower)
plt.title("Imagen original - Flower")
plt.axis("off")
plt.show()
```

Las imágenes digitales 📷 están formadas por una **matriz de píxeles**, donde cada píxel representa la intensidad de color mediante valores numéricos entre **0 y 255**. En una imagen en color, cada píxel está compuesto por tres canales —**rojo (R)**, **verde (G)** y **azul (B)**— que combinados generan la gama completa de colores visibles. Esta representación matricial hace posible convertir las imágenes en datos numéricos, permitiendo que los algoritmos y modelos de aprendizaje automático puedan analizarlas, extraer patrones y aprender de ellas. 🔢🎨


```python
# 🔹 2. Convertir la imagen a objeto PIL y explorarla
img_pil = Image.fromarray(flower)
print("Formato:", img_pil.format)
print("Tamaño (ancho, alto):", img_pil.size)
print("Modo de color:", img_pil.mode)

# Mostrar la imagen desde PIL
img_pil.show()
```

```python
# 🔹 3. Convertir la imagen a arreglo NumPy (matriz RGB)
img_array = np.array(img_pil)
print("Forma del array:", img_array.shape)

# Visualizar el array de forma parcial
print("Valores RGB de los primeros 3 píxeles:")
print(img_array[0, :3, :])  # fila 0, primeros 3 píxeles, todos los canales
```

```python
# 🔹 4. Separar y visualizar los canales de color
R = img_array[:, :, 0]  # Canal rojo
G = img_array[:, :, 1]  # Canal verde
B = img_array[:, :, 2]  # Canal azul

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
axes[0].imshow(R, cmap='Reds')
axes[0].set_title("Canal Rojo")
axes[1].imshow(G, cmap='Greens')
axes[1].set_title("Canal Verde")
axes[2].imshow(B, cmap='Blues')
axes[2].set_title("Canal Azul")

for ax in axes:
    ax.axis("off")
plt.show()
```

```python
# 🔹 5. Manipulación simple: convertir a escala de grises
gray = img_pil.convert("L")

plt.imshow(gray, cmap='gray')
plt.title("Imagen en escala de grises")
plt.axis("off")
plt.show()

# También puedes observarla como matriz:
np.array(gray)[:5, :5]  # muestra los primeros valores de intensidad
```

## Modelo de clasificación de imagenes

### Redes convolucionales

Las **redes neuronales convolucionales (CNN)** 🧠👁️ son un tipo de modelo especialmente diseñado para procesar datos visuales. En lugar de analizar la imagen completa píxel por píxel, utilizan **filtros o kernels** que se deslizan sobre ella en un proceso llamado *convolución*, extrayendo patrones locales como bordes, texturas o formas. Cada filtro genera un **mapa de características (feature map)** que resalta distintos aspectos de la imagen. Luego, etapas de **submuestreo (pooling)** reducen la dimensión de los datos conservando la información más relevante. En las capas finales, los valores obtenidos se conectan a neuronas totalmente conectadas (*fully connected layers*) que combinan toda la información para generar una predicción o clasificación. En conjunto, estas etapas permiten que el modelo aprenda jerárquicamente desde detalles simples hasta estructuras complejas dentro de una imagen. 🤖📸

### MNIST

El conjunto de datos **MNIST** 🧮 es uno de los más conocidos y utilizados en el campo del aprendizaje automático y la visión artificial. Contiene **70,000 imágenes en escala de grises** de dígitos escritos a mano (del 0 al 9), cada una con un tamaño de **28×28 píxeles**. Este dataset se ha convertido en un estándar educativo y de evaluación para probar modelos de clasificación de imágenes, ya que ofrece un problema simple pero suficientemente representativo para comprender el funcionamiento de las redes neuronales convolucionales. Gracias a su formato limpio y equilibrado, MNIST es ideal para aprender a preprocesar imágenes, entrenar modelos y medir su capacidad de generalización. ✍️🔢


```python
# Cargar el dataset MNIST
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Tamaño del conjunto de entrenamiento:", X_train.shape)
print("Tamaño del conjunto de prueba:", X_test.shape)
```

```python
# Visualizar algunos ejemplos del dataset
plt.figure(figsize=(8, 3))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(X_train[i], cmap='gray')
    plt.title(f"Etiqueta: {y_train[i]}")
    plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
#Visualizar por número 
number=6
idxs = [i for i,y in enumerate(y_train) if y==number]
plt.figure(figsize=(8, 3))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(X_train[idxs[i]], cmap='gray')
    plt.title(f"Etiqueta: {number}")
    plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# Normalizar y redimensionar las imágenes
# CNN espera entradas de 4 dimensiones: (n_imágenes, alto, ancho, canales)

X_train = X_train.reshape(-1, 28, 28, 1).astype("float32") / 255.0
X_test = X_test.reshape(-1, 28, 28, 1).astype("float32") / 255.0
```

**TensorFlow** y **Keras** 🧠⚙️ son dos de las bibliotecas más populares para el desarrollo de modelos de aprendizaje profundo. **TensorFlow**, creada por Google, es una poderosa plataforma de cómputo numérico que permite construir, entrenar y desplegar redes neuronales de manera eficiente, aprovechando tanto CPU como GPU. **Keras**, por su parte, es una interfaz de alto nivel que se integra con TensorFlow y simplifica enormemente la creación de modelos gracias a su sintaxis clara y modular. En conjunto, ambas herramientas facilitan el desarrollo de proyectos de inteligencia artificial, desde prototipos educativos hasta aplicaciones de visión artificial, procesamiento de lenguaje natural y sistemas predictivos a gran escala. 🚀🤖


#### Diseño y preparación de la red

  <img src="https://cdn.hashnode.com/res/hashnode/image/upload/v1722200188744/7666f407-bda6-4522-92ef-226e34de051c.png" alt="Red CNN" style="max-width:50%; height:auto;" width="40%">
</center>

```python
#Crear el modelo CNN con Keras
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')  # 10 clases (dígitos 0–9)
])

model.summary()
```

**Estructura de la red**


*Entrada:*  
Las imágenes de entrada son en **escala de grises de 28×28 píxeles** con un solo canal (1).  


*Bloque 1: detección de patrones simples*  
- `Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1))`  
  Aplica **32 filtros 3×3** que se deslizan por la imagen buscando patrones como bordes y líneas.  
  La activación **ReLU** deja en cero los valores negativos, ayudando al modelo a aprender de manera más estable.  

- `MaxPooling2D((2,2))`  
  Toma el valor máximo de cada bloque 2×2, reduciendo la imagen a la mitad en cada dimensión.  
  Esto se llama **submuestreo** o *pooling*, y permite conservar lo más importante mientras se reducen los datos.  

*Bloque 2: detección de patrones más complejos*  
- `Conv2D(64, (3,3), activation='relu')`  
  Añade **64 filtros** que combinan los patrones previos (bordes, esquinas, trazos curvos).  
- `MaxPooling2D((2,2))`  
  Nuevamente se reduce la imagen a la mitad, enfocándose en las características más relevantes.  

*Conversión a vector:*  
- `Flatten()`  
  Convierte la imagen 3D (de tamaño 5×5×64) en un vector de **1600 valores**.  
  Este paso permite conectar la información visual con las capas totalmente conectadas (*fully connected*).  

*Capas densas (clasificación):*  
- `Dense(64, activation='relu')`  
  Combina las características detectadas para formar patrones más abstractos.  
- `Dense(10, activation='softmax')`  
  Produce **10 probabilidades**, una por cada dígito del 0 al 9.  
  La activación *softmax* normaliza los valores para que sumen 1, eligiendo la clase con mayor probabilidad como salida final.  

*En resumen:*  
1. Las **capas convolucionales** actúan como filtros que detectan patrones visuales.  
2. Las **capas de pooling** reducen la complejidad de los datos.  
3. La **capa flatten** transforma los mapas de características en un vector.  
4. Las **capas densas** toman decisiones basadas en las características extraídas.  

💡 *Analogía:* Las convoluciones son como pequeñas lupas que exploran la imagen buscando rasgos, el pooling es como alejarse para ver el conjunto más claro, y las capas densas son el “jurado” que decide qué número representa la imagen. 🔢🤖


#### Entrenamiento

```python
# Compilar y entrenar el modelo
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

history = model.fit(X_train, y_train, epochs=10,
                    validation_data=(X_test, y_test))
```

#### Evaluación del aprendizaje

```python
#Evaluar el modelo
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=2)
print(f"Exactitud en el conjunto de prueba: {test_acc:.4f}")
```

```python
#Visualizar la evolución del entrenamiento
plt.plot(history.history['accuracy'], label='Entrenamiento')
plt.plot(history.history['val_accuracy'], label='Validación')
plt.xlabel('Época')
plt.ylabel('Exactitud')
plt.legend()
plt.title('Evolución del desempeño')
plt.show()
```

```python
#Obtener predicciones sobre el conjunto de prueba
y_pred = np.argmax(model.predict(X_test), axis=1)

# matriz de confusión
cm = confusion_matrix(y_test, y_pred)
cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]  # normalizada por clase

plt.figure(figsize=(8,6))
sns.heatmap(
    cm,
    annot=True,
    fmt=".0f",
    cmap="Blues",
    cbar=True,
    xticklabels=[str(i) for i in range(10)],
    yticklabels=[str(i) for i in range(10)],
    linewidths=1,
    linecolor='white'
)

#Etiquetas y estilo
plt.title("📊 Matriz de Confusión Normalizada - MNIST", fontsize=14, weight='bold')
plt.xlabel("Predicción del modelo", fontsize=12)
plt.ylabel("Etiqueta verdadera", fontsize=12)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.tight_layout()
plt.show()
```

#### Probar el modelo

```python
def predecir_digito(image):
    try:
        if image is None:
            return "Por favor dibuja un número 🙃"

        # Gradio Sketchpad devuelve un diccionario con la clave 'composite'
        if isinstance(image, dict):
            img_array = image.get("composite", None)
            if img_array is None:
                return "⚠️ No se pudo obtener la imagen del input ('composite' vacío)."
        else:
            return f"⚠️ Tipo de entrada no soportado: {type(image)}"

        # Convertir a escala de grises y adaptar a MNIST
        img_gray = np.array(img_array).astype(np.uint8)
        img_gray = 255 - img_gray  # Invertir colores
        img = Image.fromarray(img_gray).resize((28, 28))
        arr = np.array(img) / 255.0
        arr = arr.reshape(1, 28, 28, 1)

        # Predicción
        pred = np.argmax(model.predict(arr))
        return f"🧠 El modelo piensa que tu número es {pred}"

    except Exception as e:
        import traceback
        return f"⚠️ Error: {str(e)}\n\n{traceback.format_exc()}"
```

```python
iface = gr.Interface(
    fn=predecir_digito,
    inputs=gr.Sketchpad(
        crop_size=(256, 256),
        type="numpy",
        image_mode="L",   # Escala de grises
        brush=gr.Brush(), # Permite dibujar
        label="Dibuja un número (0–9)"
    ),
    outputs=gr.Textbox(),
    title="🖍️ Dibuja un número y deja que la red lo identifique",
    description="Red convolucional simple entrenada con MNIST."
)

iface.launch(inline=True, debug=True)
```
