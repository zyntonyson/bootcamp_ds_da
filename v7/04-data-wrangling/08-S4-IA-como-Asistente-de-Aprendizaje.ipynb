## 📚 *Durante el estudio en la plataforma*

---

**🎯 Escenario:** Explicación más simple de conceptos complejos  
💬 *Prompt sugeridos:*  

> "Explícame en palabras simples qué hace `groupby().agg()` en pandas con un ejemplo"
>> 💡 Tip: Comparaciones, analogías, ejemplos cotidianos.
    

**🎯 Escenario :** Entender errores

💬 *Prompt sugeridos:*  

![](./img/error.png)

![](./img/error_2.png)

![](./img/error_3.png)

**🎯 Escenario:** Resumen del tema  
💬 *Prompt sugerido:*  
"Resúmeme en 5 bullets las operaciones más comunes de wrangling con pandas" ( O usa el resumen del tema)

**🎯 Escenario:** Crear cuestionarios o ejercicios
💬 *Prompt sugerido:*  
- "Crea un quiz de 5 preguntas sobre filtrado de datos con `.loc`"

- "Creame un ejercicio para practicas las tablas groupby en pandas, me gustaria que sea sobre el consumo de coca-cola"

>> 💡 Tip: Pedirlos con dificultad progresiva.

## 🛠️ *Al resolver un proyecto de unidad*

---

**🎯 Escenario:** Instrucciones poco claras  
💬 *Prompt sugerido:*  ¿Qué significará duplicado engañoso?

![](./img/item_to_explain.png)

**🎯 Escenario:** Proponer plan de resolución  
💬 *Prompt sugerido:* Ayudame a resolver este ejercicio, quiero comprender lo que se me pide cómo resolverlo  

![](./img/item_to_explain_2.png)

**🎯 Escenario:** Revisión previa al envío  
💬 *Prompt sugerido:*  
"Revísame mi proyecto y dime si hay algo que mejorar antes de entregarlo"

## 🔁 *Después de entregar el proyecto*

----

**🎯 Escenario:** Entender comentarios del revisor  
💬 *Prompt sugerido:*  
"No entiendo del todo que es lo que se me pide corregir, podrias explicarme y darme un ejemplo guiado" 

![](./img/comentario_revisor.png)



**🎯 Escenario 3:** Revisión final antes de reenviar  
💬 *Prompt sugerido:*  
"¿Puedes revisar si esta nueva solución cumple con lo que pedía el revisor?"




## 🔁 *Repaso y autoevaluación*

---


**🎯 Escenario 1:** Resumen desde transcripciones de clase  
💬 *Prompt sugerido:*  
- Resumen sobre la clase
- Cuales son los temas de mayor dificultad
- Buscar momentos especificos
- Pedir cuestionario


**🎯 Escenario 3:** Quiz a partir de mis notas  

![](./img/b03dbd3955627decd35245eeb967c0a0.jpg)

💬 *Prompt sugerido:*  
"Con mis notas de pandas, hazme una trivia de 3 preguntas con opciones y la respuesta correcta"

---

✨ *Ideas clave para llevarte de esta sesión:*
1. La IA puede explicarte, retarte y corregirte como un tutor personalizado 24/7 🤖
2. Un buen *prompt* es claro, con contexto y pide justo lo que necesitas
3. No se trata de que la IA lo haga por ti, sino de que aprendas más rápido y con más confianza 💪
4. Tu pensamiento crítico y práctica siguen siendo el ingrediente principal del aprendizaje 🧠

¡Haz de la IA tu aliada para convertirte en un gran analista de datos! 🐼🚀


>> “La IA no viene a hacer tu trabajo, viene a multiplicar tu capacidad de aprender y crear 🚀”
