## 🌿 1. Git & GitHub: Tu Máquina del Tiempo y tu Vitrina Virtual

Antes de tocar el código, aclaremos dos conceptos fundamentales que a menudo se confunden, pero que trabajan en equipo:

*   **Git (El Motor 🚗):** Es un sistema de control de versiones que corre localmente en tu computadora. Imagina que es una "máquina del tiempo" o una cámara de fotos. Cada vez que haces un cambio importante en tu código, Git toma una foto (llamada *commit*). Si algo se rompe en el futuro, puedes regresar en el tiempo a cualquier foto anterior. ¡Se acabaron los archivos llamados `proyecto_final_v2_este_si_definitivo.py`!
*   **GitHub (La Nube ☁️):** Es una plataforma web que hospeda tus proyectos que usan Git. Es como el "Instagram" o el "Dropbox" de los desarrolladores. Te permite almacenar tus códigos de forma segura, colaborar con otros equipos y mostrar tus proyectos como un portafolio profesional frente a reclutadores.

## 📂 2. Preparando el Repositorio y el Entorno de Trabajo



¡Manos a la obra! Vamos a crear nuestro espacio en la nube y a traerlo a nuestra computadora.

### Paso 1: Crea tu cuenta en GitHub
Si aún no la tienes, ve a [github.com](https://github.com) y regístrate. Es completamente gratis y solo requiere un correo electrónico.

### Paso 2: Crea un Repositorio Público en GitHub
1. Inicia sesión en tu cuenta de GitHub.
2. En la esquina superior derecha, haz clic en el botón **"+"** y selecciona **New repository (Nuevo repositorio)**.
3. Configura los siguientes campos:
   * **Repository name:** `mi-primer-deploy` (usa minúsculas y guiones o pon el nombre que quieras darle a tu proyecto).
   * **Description:** `Mi primera aplicación de datos interactiva desplegada en Render`.
   * **Public:** Asegúrate de que esté marcado como **Público** (necesario para que Render pueda acceder a él de forma gratuita).
   * **Add a README file:** Marca esta casilla (creará un archivo inicial para describir el proyecto).

4. Haz clic en el botón verde **Create repository**.




### Paso 3: Clona el Repositorio en tu Computadora usando VS Code
No necesitas usar la consola ni comandos complejos para traer tu proyecto a local. Lo haremos directo desde VS Code:

1. Abre **Visual Studio Code**.
2. Abre la **Paleta de Comandos** presionando las teclas:
   * **Windows/Linux:** `Ctrl + Shift + P`
   * **macOS:** `Cmd + Shift + P`
3. Escribe `Git: Clone` y presiona **Enter**.
4. Selecciona la opción **Clone from GitHub (Clonar desde GitHub)**.
   > 🔑 *Nota:* Si es la primera vez, VS Code abrirá una ventana en tu navegador pidiendo autorización para conectarse a tu cuenta de GitHub. Haz clic en autorizar.
5. Verás una lista de tus repositorios. Selecciona `mi-primer-deploy` (o el nombre que le hayas asignado).
6. VS Code te pedirá elegir una carpeta en tu computadora donde guardar el proyecto. Elige una ubicación cómoda (como tu Escritorio o tu carpeta de Documentos).
7. Al finalizar la descarga, aparecerá un mensaje en la esquina inferior derecha. Haz clic en **Open (Abrir)** para abrir tu nuevo espacio de trabajo.

## 🤖 3. Generando la Aplicación Streamlit con Apoyo de IA

Una vez dentro de la carpeta clonada en VS Code, estructuraremos nuestro proyecto. Crea la carpeta `data/` y la carpeta `prompts/` en la raíz de tu proyecto. El orden visual debe verse así:


```text
mi-primer-deploy/
├── README.md          # Archivo de texto creado por GitHub
├── data/              # Carpeta para almacenar el archivo CSV de datos
├── prompts/           # Carpeta para registrar tus prompts
├── app.py             # Archivo principal de nuestra aplicación interactiva
└── requirements.txt   # Listado de librerías para el servidor
```

### 📊 Parte A: Generación de Datos Sintéticos (CSV)
Usaremos Inteligencia Artificial para crear nuestro conjunto de datos ficticio adaptado al negocio que queramos simular.

Crea un archivo llamado `prompts/generar_datos.txt` y escribe en él la instrucción que enviarás a la IA. Aquí tienes una plantilla robusta y personalizable:


```text
Actúa como un analista de datos experto en búsqueda y preparación de datos  que  genere un conjunto de datos sintético en formato CSV.
La temática del conjunto de datos debe ser sobre: [PERSONALIZA AQUÍ LA TEMÁTICA, EJEMPLO: Ventas de una cafetería gourmet especializada].
el conjunto de datos  debe cumplir con las siguientes directrices:
1. Generar exactamente [150] filas de datos aleatorios pero realistas.
2. Contener obligatoriamente las siguientes columnas categoricas, numericas, de fecha y boolenas
3. Genera los datos separados por comas 
```

Guarda los datos de la salida en el archivo `data/dataset.csv`

---

### 🎨 Parte B: Creación de la Aplicación con Streamlit y Plotly


Ahora diseñaremos el dashboard que verán los usuarios en internet. Utilizaremos **Plotly** para gráficos interactivos y **Streamlit** como la estructura web.

Crea un archivo llamado `prompts/generar_app.txt` y copia en él esta instrucción para la IA:

```text
Actúa como un desarrollador Frontend de datos experto en Python, Streamlit y Plotly Express. Escribe el código completo para `app.py` que cargue el conjunto de datos de "data/dataset.csv" y muestre un panel de control interactivo.
La aplicación debe incluir lo siguiente:

1. Cargar los datos desde la ruta `data/dataset.csv`
2. Un título y un subtítulo muy llamativos con emojis, acordes a la temática de los datos.

3. Un checkbox para mostrar/ocultar una vista previa interactiva del DataFrame filtrado (mostrando las primeras 5 a 10 filas usando st.dataframe).

6. Tres gráficos altamente interactivos creados con Plotly Express que se actualicen al cambiar los filtros:
   - Gráfico 1: Un gráfico de barras horizontales o verticales.
   - Gráfico 2: Un gráfico circular (Pie chart / Donut chart).
   - Graifoc 3: Un histograma
7. Los gráficos deben tener un diseño estético profesional y usar el ancho completo de la columna donde se ubiquen.
```

Crea el archivo `app.py` en la raíz de tu proyecto y guarda el código generado por la IA. 


---

### 📦 Parte C: Definir los Requerimientos (`requirements.txt`)


Para que un servidor en internet sepa qué librerías de Python debe instalar para que nuestro código funcione, creamos el archivo `requirements.txt`.

Crea el archivo `requirements.txt` en la raíz de tu proyecto e ingresa el siguiente contenido:

```text
streamlit
plotly
pandas
```

> ⚠️ **Importante:** No coloques números de versión en el archivo `requirements.txt` (por ejemplo, escribe `streamlit` y no `streamlit==1.25.0`). Al omitirlas, Render instalará automáticamente las versiones más recientes y estables compatibles con sus sistemas operativos Linux de producción, evitando conflictos de dependencias comunes.

### ⚡ Prueba Local de tu Aplicación
Abre la terminal integrada de VS Code (`Ctrl + ~` o ve al menú superior `Terminal -> New Terminal`) y corre los siguientes comandos:

```bash
# 1. Instala las librerías necesarias
pip install -r requirements.txt

# 2. Arranca el servidor local de Streamlit
streamlit run app.py
```

Se abrirá automáticamente una ventana en tu navegador web mostrando tu aplicación corriendo en la dirección local `http://localhost:8501`. Interactúa con los filtros y comprueba que los gráficos interactivos de Plotly respondan al instante. Una vez verificado que todo funciona correctamente, detén el servidor presionando `Ctrl + C` en tu terminal de VS Code.

## 📤 4. Carga de Cambios en GitHub (Versionado)




Ahora subiremos todo nuestro proyecto local a nuestro repositorio de GitHub en internet. 

Asegúrate de ejecutar los siguientes tres comandos de Git en tu terminal de VS Code. ¡Mira lo fácil que es entenderlos con esta analogía de viaje! ✈️

1. **`git add .` (Hacer la maleta 🧳):** Este comando le dice a Git: *"Busca todos los archivos nuevos y modificados de mi carpeta y colócalos en la zona de preparación"*.
   ```bash
   git add .
   ```
2. **`git commit -m "Mensaje descriptivo"` (Ponerle etiqueta al equipaje 🏷️):** Toma una foto instantánea con una breve etiqueta que describa qué hiciste.
   ```bash
   git commit -m "Aplicación de streamlit"
   ```
3. **`git push` (Enviar el equipaje al avión 🚀):** Sube todas las fotos y archivos guardados de tu computadora local al repositorio remoto en los servidores de GitHub.
   ```bash
   git push
   ```

*Entra a tu página de GitHub y refresca tu repositorio. ¡Allí verás listados tus archivos `app.py`, `requirements.txt` y la carpeta `data/`!*

### 📝 Diseño del README.md Profesional



Para que tu repositorio no tenga una página en blanco y luzca excelente para cualquiera que te visite, escribe un prompt para generar una presentación detallada de tu proyecto. 

Crea un archivo llamado `prompts/prompt_readme.txt` y copia en él esta instrucción para la IA:

```text
Actúa como un redactor técnico y especialista en marcas de portafolios de datos. Escribe un archivo README.md espectacular y en español para mi proyecto en GitHub. 
El proyecto es una aplicación web interactiva que analiza  [PERSONALIZA AQUÍ CON TU TEMÁTICA, EJEMPLO: Cafetería Gourmet] usando Python, Streamlit y gráficos interactivos con Plotly Express.
El README debe incluir las siguientes secciones con emojis llamativos:
1. Título del proyecto y descripción introductoria.
3. Instrucciones paso a paso para la instalación en local (clonación del repositorio, instalación de dependencias en requirements.txt y comando de inicio streamlit run).
4. Sección indicando la tecnología utilizada.
El tono debe ser motivador y profesional. Utiliza formato markdown enriquecido (listas ordenadas, bloques de código, negritas y citas).
```

## 🚀 5. Despliegue de la Aplicación en Render



¡El paso definitivo! Haremos que tu aplicación web sea accesible desde cualquier parte del mundo con un enlace único y gratuito.

1. **Inicia sesión en Render:** Ve a [render.com](https://render.com) y haz clic en **Sign In**. Selecciona el botón **GitHub** para ingresar usando las mismas credenciales (esto vinculará tus cuentas automáticamente y hará todo más fácil).
2. **Crea un Nuevo Servicio Web:** En la parte superior de tu panel de Render (Dashboard), haz clic en el botón azul **New +** y selecciona la opción **Web Service**.
3. **Conecta tu Repositorio:**
   * En el listado de repositorios de GitHub que aparecerá en pantalla, busca tu repositorio `mi-primer-deploy`.
   * Haz clic en el botón **Connect** a su derecha.
4. **Configura el Despliegue:** Completa los siguientes datos en el formulario:
   * **Name:** `mi-cafeteria-dashboard` (o el nombre único que desees para tu URL).
   * **Region:** Selecciona la región predeterminada o la más cercana a ti (ej. *Oregon* o *Frankfurt*).
   * **Branch:** `main` (o la rama principal de tu repositorio).
   * **Runtime:** `Python` (o `Python 3`).
   * **Build Command (Comando de construcción):**
     ```bash
     pip install -r requirements.txt
     ```
   * **Start Command (Comando de inicio):**
     ```bash
     streamlit run app.py 
     ```

   * **Instance Type (Tipo de instancia):** Asegúrate de que esté seleccionada la opción **Free** ($0 / month).
5. **Crea el Servicio:** Desplázate hasta el final del formulario y haz clic en el botón **Deploy Web Service** (o **Create Web Service**).
6. **Espera el proceso "In-Live":**
   * Render comenzará a construir tu aplicación: creará el contenedor Linux, instalará Python, descargará las dependencias de tu archivo `requirements.txt` y correrá el comando de inicio de Streamlit.
   * Puedes ver el avance en la consola de texto en vivo en Render. Este primer despliegue suele tardar de **2 a 4 minutos**.
   * Una vez finalizado el proceso de instalación, verás un mensaje que dice: `Your service is live 🎉`.
7. **Visita tu aplicación pública:**
   * En la esquina superior izquierda de tu pantalla de Render verás un enlace en color azul con formato: `https://tu-nombre-de-app.onrender.com`.
   * Haz clic sobre él... **¡y listo!** Tu aplicación ya se encuentra activa en internet. Comparte ese enlace con tus amigos, familiares o en tu perfil de LinkedIn. 🚀🥳
