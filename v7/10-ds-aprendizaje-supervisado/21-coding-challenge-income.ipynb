### SMOTE

⚖️ *Balanceo de clases con SMOTE (Oversampling sintético)*

Cuando la clase minoritaria (por ejemplo, `>50K`) tiene muchas menos observaciones que la clase mayoritaria (`<=50K`), los modelos tienden a **aprender peor los patrones de la clase minoritaria**.  
Una técnica popular para mitigar este problema es **SMOTE**, que genera ejemplos sintéticos de la clase menos representada.

---

```python
# Instalamos la librería si no está disponible
# !pip install imbalanced-learn

from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import pandas as pd

# Supongamos que ya tenemos un DataFrame df con las variables preparadas
# X = variables predictoras, y = variable objetivo codificada (0 = <=50K, 1 = >50K)

# Ejemplo simplificado
X = df.drop(columns=['income'])
y = df['income'].apply(lambda x: 1 if x == '>50K' else 0)

# División de datos antes del oversampling
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print('Distribución original en entrenamiento:')
print(y_train.value_counts(normalize=True))

# Aplicación de SMOTE
smote = SMOTE(random_state=42, sampling_strategy='auto')
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

print('\nDistribución después del oversampling:')
print(y_train_res.value_counts(normalize=True))
