## ✨ *¿Que és la Optimización?*  

La optimización es el proceso de *encontrar la mejor solución posible* dentro de un conjunto de alternativas, respetando ciertas restricciones. En términos simples, optimizar significa *maximizar o minimizar* una función objetivo: por ejemplo, maximizar las ganancias de una empresa o minimizar los costos de producción.  

En la vida diaria encontramos múltiples ejemplos:  
- Elegir la ruta más corta al conducir 🚗  
- Distribuir un presupuesto para obtener mayor beneficio 💰  
- Diseñar horarios eficientes ⏰  

Existen dos enfoques principales:  
* **Optimización exacta**: busca la solución perfecta (ej. resolver una ecuación con un método algebraico).  
* **Optimización aproximada**: se centra en encontrar soluciones suficientemente buenas cuando la exacta es muy costosa o imposible de calcular (ej. usar algoritmos iterativos como el descenso del gradiente).  


## 🚀 Descenso del Gradiente  

El **descenso del gradiente** es un algoritmo iterativo de optimización fundamental en *Machine Learning*. Su objetivo es encontrar los valores de parámetros que **minimizan una función de costo** (por ejemplo, el error entre predicciones y valores reales).  

La idea es simple: calcular la pendiente de la función y moverse en la dirección contraria, dando pasos controlados por una tasa de aprendizaje $\alpha$:  

$$
\theta = \theta - \alpha \nabla J(\theta)
$$  

Donde:  
- $\theta$ son los parámetros del modelo.  
- $\alpha$ es la tasa de aprendizaje.  
- $\nabla J(\theta)$ es el gradiente de la función de costo.  

👉 Su importancia radica en que es la base para entrenar modelos como regresión lineal, regresión logística y redes neuronales, permitiendo ajustar los parámetros hasta lograr predicciones más precisas.  


```python

# Función y su derivada
def f(x): return x**2
def df(x): return 2*x

# Inicialización
x = -10
alpha = 0.1 # tasa de aprendizaje
x_hist = [x]

# Iteraciones
for _ in range(20):
    grad = df(x)
    x = x - alpha * grad
    x_hist.append(x)

# Graficar
x_vals = np.linspace(-12, 12, 100)
plt.plot(x_vals, f(x_vals), label='f(x) = x²')
plt.plot(x_hist, f(np.array(x_hist)), 'ro-', label='Pasos del descenso')
plt.title("Descenso de gradiente en f(x) = x²")
plt.xlabel("x"); plt.ylabel("f(x)")
plt.legend()
plt.grid()
plt.show()
```

### *Entrenamiento de la regresion lineal por descenso de gradiente*

El descenso del gradiente estima las betas ajustándolas paso a paso para minimizar la función de costo, que mide los errores de predicción entre valores observados y estimados, logrando así un modelo de regresión más preciso y eficiente.

```python
# Datos simulados
np.random.seed(0)
obs =200
X = 2 * np.random.rand(obs, 1)  # Variable independiente
y = 4 + 3 * X + np.random.randn(obs, 1)  # Etiquetas con ruido
```

```python
plt.scatter(X, y, color="blue", alpha=0.7, edgecolor="k")
plt.xlabel("X (Variable independiente)")
plt.ylabel("y (Variable dependiente)")
plt.title("Datos simulados: dispersión de X vs y")
plt.show()
```

```python
# Añadir columna de unos para el término de sesgo (intercepto)
X_b = np.c_[np.ones((obs, 1)), X]  
```

En este caso, la función de costo utilizada es el **Error Cuadrático Medio (MSE)**, que mide la diferencia promedio al cuadrado entre los valores reales y los predichos por el modelo. Su forma es:  

$$
\mathrm{MSE}(\beta)= \frac{1}{m}\varepsilon^\top\varepsilon= \frac{1}{m}(y - X\beta)^\top(y - X\beta),
$$
donde $X\in\mathbb{R}^{m\times p}$ es la matriz de diseño, $y\in\mathbb{R}^m$ es el vector de respuestas y $\varepsilon=y-X\beta$ es el residuo.


Expandimos $\mathrm{SSE}(\beta)$ usando álgebra matricial:
$$
\begin{aligned}
\mathrm{MSE}(\beta)
&=\frac{1}{m}(y-X\beta)^\top(y-X\beta)\\
&=\frac{1}{m}(y^\top y - 2\,y^\top X\beta + \beta^\top X^\top X \beta).
\end{aligned}
$$

Derivamos con respecto a $\beta$ y simplificamos:
$$
\nabla_\beta \mathrm{SSE}(\beta)= \frac{1}{m}(-2X^\top y + 2X^\top X \beta) \Longrightarrow\;\;
\frac{2}{m} X^T \left( X\beta - y \right).
$$





👉 Esta expresión es la que se implementa en el código para actualizar iterativamente los parámetros y minimizar el error.  


```python
# Inicialización aleatoria de los parámetros (theta0 y theta1)
theta = np.random.randn(2, 1)
alpha = 0.1  # Tasa de aprendizaje
n_iterations = 1000
m = len(X_b)

# Algoritmo de descenso de gradiente
for iteration in range(n_iterations):
    gradients = 2/m * X_b.T.dot(X_b.dot(theta) - y)  # Derivada de la función de costo
    theta = theta - alpha * gradients  # Actualización de parámetros

print("Theta final (intercepto y pendiente):", theta.ravel())
```

```python
p,m_ = theta.ravel()
x = np.linspace(min(X), max(X), 100)  # de 0 a 2 con 100 puntos
y_pred = m_ * x + p
plt.scatter(X, y, color="blue", alpha=0.7, edgecolor="k")
plt.plot(x, y_pred, color="red", label=f"y = {m_:.2f}x + {p:.2f}")
plt.xlabel("X (Variable independiente)")
plt.ylabel("y (Variable dependiente)")
plt.title("Datos simulados: dispersión de X vs y")
plt.legend()
plt.show()
```

### ⚡ Descenso de Gradiente Estocástico (SGD)

El **descenso de gradiente estocástico (SGD)** es una variante del descenso de gradiente tradicional que actualiza los parámetros utilizando **una muestra aleatoria** (o un pequeño subconjunto) en cada iteración.  

👉 **Concepto:** En lugar de calcular el gradiente completo con todo el dataset (lo cual es costoso cuando hay millones de datos), se aproxima el gradiente usando solo una muestra.  

👉 **Intuición:** Es como si quisieras bajar una montaña a oscuras, pero en lugar de calcular la pendiente de toda la superficie, das un paso usando solo la información de un punto cercano. Puede ser un poco más ruidoso, pero te permite avanzar mucho más rápido.  

👉 **Mejora sobre el gradiente tradicional:**  
- El **gradiente batch** (clásico) requiere recorrer todos los datos en cada iteración:  

$$
\nabla J(\beta) = \frac{2}{m} X^T (X\beta - y)
$$  

- El **SGD** usa solo una muestra $(x^{(i)}, y^{(i)})$:  

$$
\nabla J(\beta) \approx 2 \cdot x^{(i)} \left( x^{(i)T}\beta - y^{(i)} \right)
$$  

Esto introduce ruido, pero acelera el cálculo y ayuda a escapar de mínimos locales.  


```python
# Inicialización de parámetros
theta = np.random.randn(2, 1)  # [intercepto, pendiente]
n_epochs = 50
t0, t1 = 5, 50  # parámetros para tasa de aprendizaje adaptativa

# Función para decrecer la tasa de aprendizaje con el tiempo
def learning_rate_schedule(t):
    return t0 / (t + t1)

# Descenso de Gradiente Estocástico
for epoch in range(n_epochs):
    for i in range(m):
        random_index = np.random.randint(m)          # selecciona un índice aleatorio
        xi = X_b[random_index:random_index+1]        # una muestra aleatoria
        yi = y[random_index:random_index+1]          # valor objetivo correspondiente
        gradients = 2 * xi.T.dot(xi.dot(theta) - yi) # gradiente usando una muestra
        alpha = learning_rate_schedule(epoch * m + i)
        theta = theta - alpha * gradients            # actualización de parámetros

print("Theta final por SGD:", theta.ravel())
```

## 🔒 Regularización en la Regresión

La **regularización** es una técnica utilizada en modelos de regresión para evitar el **sobreajuste (overfitting)**.  
En términos simples, añade una **penalización** al crecimiento de los coeficientes, evitando que el modelo se ajuste demasiado al ruido de los datos.  

👉 **Intuición:**  
- Sin regularización, la regresión puede inflar algunos coeficientes para reducir el error de entrenamiento.  
- La regularización actúa como una “restricción” que mantiene los parámetros bajo control.  
- Imagina que cada coeficiente tiene una liga o un imán que lo jala hacia cero.  

---

### 📌 Ridge Regression (L2)

En **Ridge**, la penalización es el **cuadrado de los coeficientes**.  
Esto “achica” los parámetros, pero rara vez los lleva exactamente a cero.  

La función de pérdida cambia de:  

$$
J(\beta) =\frac{1}{m}(y-X\beta)^\top(y-X\beta)\\
$$  

a:  

$$
J_{Ridge}(\beta) = \frac{1}{m}(y-X\beta)^\top(y-X\beta) +  \lambda\|\beta\|_2^2
$$  

👉 Intuición: como ponerle una **liga elástica** a los coeficientes para que no crezcan demasiado.  

---

### 📌 Lasso Regression (L1)

En **Lasso**, la penalización es la **suma del valor absoluto** de los coeficientes.  
Esto no solo reduce su magnitud, sino que puede hacer que algunos coeficientes se vuelvan **exactamente cero**, funcionando también como un método de selección de variables.  

La función de pérdida es:  


$$
J_{Lasso}(\beta) = \frac{1}{m}(y-X\beta)^\top(y-X\beta) +  \lambda\|\beta\|
$$  

👉 Intuición: como si un **imán** atrajera los coeficientes hacia cero, apagando algunos por completo.  

---

### ✨ Diferencias clave

- **Ridge (L2):** achica todos los coeficientes, ninguno queda en cero.  
- **Lasso (L1):** puede eliminar variables, dejando coeficientes exactamente en cero.  

> En la práctica, elegir entre Ridge o Lasso depende de si queremos solo **reducir la varianza** (Ridge) o además **seleccionar variables** (Lasso).


```python
# Entrenamiento del modelo con regularización L2
ridge_reg = Ridge(alpha=1, solver="cholesky")  # alpha controla la fuerza de la regularización
ridge_reg.fit(X, y)
print("Coeficientes con Ridge:", ridge_reg.intercept_, ridge_reg.coef_)
```

```python
# Entrenamiento del modelo con regularización L1
lasso_reg = Lasso(alpha=0.1)  # alpha controla la fuerza de la regularización
lasso_reg.fit(X, y)
print("Coeficientes con Lasso:", lasso_reg.intercept_, lasso_reg.coef_)
```

## 🧠 Redes Neuronales: Intuición y Conceptos Básicos



Las **redes neuronales artificiales** son modelos inspirados en el cerebro humano, diseñados para reconocer patrones y aprender de los datos.  

👉 **Intuición:**  
Así como una neurona biológica recibe señales, las procesa y emite una salida, una neurona artificial recibe **entradas numéricas**, aplica un **peso** a cada una, las suma, pasa el resultado por una **función de activación** y genera una salida.

### 🏗️ Estructura básica
Una red neuronal se organiza en **capas**:  
- *Capa de entrada*: recibe las variables de los datos.  
- *Capas ocultas*: transforman la información a través de múltiples neuronas.  
- *Capa de salida*: produce la predicción final (ej. probabilidad de churn, clasificación de imágenes).  

---

### ⚙️ ¿Cómo se entrenan?
El entrenamiento busca ajustar los **pesos ($w$)** y los **sesgos ($b$)** para que la salida de la red se acerque a la realidad.  
El proceso se basa en dos pasos:  
1. **Forward propagation**: los datos recorren la red y producen una predicción.  
2. **Backward propagation**: se calcula el error y se propaga hacia atrás, ajustando pesos.
