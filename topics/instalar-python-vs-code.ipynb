# 🐍📊 Cómo tener un setup básico de Python para análisis de datos

Hemos trabajado principalmente en **Google Colab**, es una **habilidad muy deseable** como analistas de datos poder ejecutar nuestros análisis **desde nuestra propia computadora** 💻.  

En este webinar veremos **cómo crear un setup básico, funcional y realista de Python**, suficiente para trabajar con scripts, notebooks y librerías de análisis de datos en un entorno local.



## 🛠️ Herramientas 

### 🖥️ Terminal

La **terminal** nos permite comunicarnos directamente con el sistema operativo usando comandos de texto.

Como analistas de datos, usar la línea de comandos nos ayuda a:
- Ejecutar scripts de Python
- Instalar librerías
- Navegar carpetas de proyectos
- Automatizar tareas


### 🧩 VS Code
**Visual Studio Code (VS Code)** es un editor de código ligero pero muy poderoso.

Ventajas clave:
- Soporte excelente para Python
- Integración con terminal
- Autocompletado y detección de errores
- Permite trabajar con scripts y notebooks en un solo lugar



### 🐍 Intérprete de Python
El **intérprete de Python** es el programa que:
- Lee tu código `.py`
- Lo ejecuta línea por línea
- Devuelve resultados o errores

Tener Python instalado significa que **tu computadora entiende Python** 😉



## ⚙️ Pasos para tener el setup

### 🔍 Identificar si tenemos Python instalado

En la terminal, prueba alguno de estos comandos:

```bash
py --version
```

```bash
python --version
```



```bash
python3 --version
```

Si ves algo como:

`Python 3.xx.xx`

> 👉 ¡Perfecto! Ya tienes Python instalado.



### ⬇️ Instalar Python (si no lo tienes)

Si **no aparece Python**, descárgalo según tu sistema operativo:

- **Windows**: instalador desde python.org  
  ✔️ Asegúrate de marcar *“Add Python to PATH”*
- **macOS**:  
  - Instalador oficial  
  - o usando Homebrew
- **Ubuntu / Linux**:
  
```bash
sudo apt update
sudo apt install python3 python3-pip
```



### 🧑‍💻 Descargar VS Code

1. Descarga VS Code desde su sitio oficial
2. Instálalo como cualquier programa
3. Ábrelo y revisa lo siguiente:

**Extensiones recomendadas:**
- Python
- Jupyter
- Pylance

**Configuraciones clave:**
- Abrir la terminal integrada
- Seleccionar el intérprete de Python
- Ejecutar archivos `.py` desde VS Code

> 🛠️ *VS Code será nuestro centro de trabajo.*



## 🌍 Entorno global de Python y ambientes virtuales

Cuando instalas Python, existe un **entorno global**:
- Todas las librerías se instalan ahí
- Es compartido por todos los proyectos

Los **ambientes virtuales**:
- Aíslan dependencias por proyecto
- Evitan conflictos de versiones

👉 **Importante**:  
En esta práctica **trabajaremos con el entorno global**, solo para entender el flujo completo sin complejidad extra.



## 📦 Instalación de librerías con pip

Para instalar librerías, usaremos siempre la forma segura:

```bash
python -m pip install nombre_paquete
```

### 📚 Librerías que instalaremos

```bash
python -m pip install pandas matplotlib seaborn jupyter ipykernel ipywidgets
```

Estas librerías nos permiten:
- 📊 Análisis de datos (`pandas`)
- 📈 Visualización (`matplotlib`, `seaborn`)
- 📓 Notebooks (`jupyter`)
- 🔌 Integración con VS Code (`ipykernel`, `ipywidgets`)



### 📄 Instalación desde un `requirements.txt`

Un archivo `requirements.txt` contiene las librerías del proyecto:

Ejemplo de contenido:

```plaintext

pandas
matplotlib
seaborn
jupyter
```

Para instalar todo:

```bash
python -m pip install -r requirements.txt
```

> 💡 *Esto es muy común en proyectos reales y equipos de trabajo.*



## ▶️ Ejecutar nuestro primer script en Python

### 🗂️ Crear carpeta del proyecto

Crea en tu computadora una carpeta y nómbrala como `my_python`, ábrela en *VS-code*


### 📝 Crear un archivo `main.py`

Dentro de VS Code, crea un archivo `main.py` con este contenido:



## 📓 Crear nuestro primer notebook

En la **misma carpeta**, crea un archivo `analisis.ipynb`.

### 📊 Cargar un DataFrame sencillo

```python
import pandas as pd

data = {
    "nombre": ["Ana", "Luis", "María", "Carlos"],
    "edad": [23, 31, 19, 45],
    "ventas": [250, 400, 150, 300]
}

df = pd.DataFrame(data)
df
```
