### Resumen 🔃

**pandas** es la librería fundamental de Python para manipular, limpiar y analizar datos estructurados. Su nombre proviene de *Panel Data* y está construida sobre **NumPy**, lo que le da gran velocidad y flexibilidad para trabajar con grandes volúmenes de información. Algunos métodos y atributos que vimos en la sesión anterior fueron:  

- `read_csv()` : Sirve para cargar datos desde archivos CSV.  
- `shape` : Atributo que muestra la forma del DataFrame, es decir, el número de filas y columnas.  
- `columns` : Atributo que devuelve el listado con los nombres de las columnas del DataFrame.  
- `info()` : Método que muestra un resumen general del DataFrame (tipos de datos, valores nulos, memoria usada, etc.).  
- `head()` : Método que muestra las primeras filas del DataFrame (por defecto las primeras 5).  
- `drop()` : Método que permite eliminar columnas o filas específicas.  
- `rename()` : Método que se utiliza para cambiar el nombre de columnas o índices.  
- `describe()` : Método que calcula estadísticas descriptivas básicas de las columnas numéricas del DataFrame.  
- `[[]]`: Combinación para la selección de columnas

> 💡 Estos métodos y atributos son esenciales para explorar y comprender la estructura de tus datos antes de realizar cualquier análisis más avanzado.

### 🧼 Limpieza y transformación de datos

#### **Formatear datos**


**Formatear cadenas de texto**

Veamos las necesidades de limpieza de la columna `Entity` 

```python
#Valores únicos de la columna
print(emission_co2['Entity'].nunique())
emission_co2['Entity'].unique().tolist()
```

```python
# Convertir texto a minúsculas
emission_co2["Entity"] = emission_co2["Entity"].str.lower()
```

**Pon en practica 🏋️✍️** 

Prueba los efecto de usar:

- `str.upper()` -> Convertir a mayúsculas
- `str.title()` -> Convertir primer letra a mayúscula
- `str.strip()` -> Remover espacios extras
- `str.replace()` -> Reemplazar cadenas de texto
    

```python
emission_co2["Entity"]=emission_co2["Entity"].str.strip()
emission_co2["Entity"]=emission_co2["Entity"].str.replace('"','')
emission_co2["Entity"]=emission_co2["Entity"].str.upper()
```

```python
# Tambien podemos aplicar varios cambios encadenados
emission_co2["Entity"]=emission_co2["Entity"].str.strip().str.replace('"','').str.upper()
```

```python
print(emission_co2['Entity'].nunique())
emission_co2['Entity'].unique().tolist()
```

**Convertir valores numéricos**

```python
emission_co2['Annual CO₂ emissions (per capita)'].unique().tolist()
```

```python
emission_co2["unit_2"]=emission_co2["unit_2"].str.strip().str.replace(',','.')
emission_co2["unit_2"]=pd.to_numeric(emission_co2["unit_2"],errors='coerce')
```

```python
emission_co2["Annual CO₂ emissions (per capita)"]=pd.to_numeric(emission_co2["Annual CO₂ emissions (per capita)"],errors='coerce')
emission_co2["Year"]=pd.to_numeric(emission_co2["Year"],errors='coerce')
```

```python
emission_co2.rename(columns={"Annual CO₂ emissions (per capita)":'annual_emissions','Year':'year','Entity':'entity'},inplace=True)
```

**Manejo de fechas**

```python
emission_co2["date"] = pd.to_datetime(emission_co2["date"], errors="coerce", utc=True)
```

```python
emission_co2.info()
```

```python
emission_co2.head()
```

#### **Manejo de valores nulos/faltantes**

```python
# Ver nulos en el dataset
emission_co2.info()
```

*¿Qué hacer con lo datos nulos?*

```python
# Eliminar filas con valores nulos
emission_co2.dropna(inplace=True)
```

```python
# Reemplazar valores nulos 
emission_co2["year"].fillna(1950, inplace=True) # Recuerda inplace hace el cambio sin necesida de guardar explícitamente 
```

> Podemos reemplazar los valores nulos por alguna estádisica de interés de los datos (mediana,percentil,media,etc)

```python
avg_unit_2=emission_co2["unit_2"].mean()
emission_co2["unit_2"]=emission_co2["unit_2"].fillna(avg_unit_2) # Nota que aquí no se usa el inplace
```

```python
emission_co2.info()
```

#### **Manejo de duplicados**

```python
# Detectar duplicados
emission_co2.duplicated().sum()
```

```python

# Eliminar duplicados
emission_co2.drop_duplicates(inplace=True) # Nota el uso de inplace
```

```python
emission_co2.info()
```

### 🔎 Filtrar, ordenar y resumir datos
