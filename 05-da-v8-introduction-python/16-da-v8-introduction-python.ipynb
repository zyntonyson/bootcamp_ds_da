## Qué es python 🐍 y su uso en análisis de datos



Python 🐍 es un lenguaje de programación versátil, fácil de aprender y con una sintaxis cercana al lenguaje humano, ideal para quienes inician en análisis de datos. Su fuerza está en su ecosistema de librerías como **NumPy**, **pandas**, **Matplotlib** y **scikit-learn**, que permiten desde limpiar y transformar datos hasta crear visualizaciones y modelos de *machine learning*.

> Piensa en Python como una 🧰 navaja suiza: automatiza tareas, procesa grandes volúmenes de información y convierte datos en conocimiento para la toma de decisiones.


## Pensamiento analítico / pensamiento algorítmico 🧠💡



El **pensamiento analítico** consiste en descomponer un problema en partes más pequeñas para entenderlo y encontrar patrones. En análisis de datos, esto significa observar los datos desde distintas perspectivas, buscar relaciones y proponer hipótesis con base en evidencia.

Por otro lado, el **pensamiento algorítmico** se enfoca en cómo resolver esos problemas paso a paso, mediante instrucciones claras y ordenadas —como si explicaras a una computadora qué hacer. 🧩  

> En pocas palabras: el pensamiento analítico te ayuda a **entender el problema**, y el algorítmico a **diseñar la solución**. Juntos son la base para transformar datos en conocimiento accionable.

---

### 🔄 Flujo típico de análisis de datos en Python

1. **Definición del problema:** comprender qué pregunta queremos responder o qué decisión queremos apoyar.  
2. **Obtención de datos:** cargar la información desde archivos, bases de datos o APIs.  
3. **Limpieza y transformación:** tratar valores nulos, corregir errores y preparar los datos para el análisis.  
4. **Análisis exploratorio (EDA):** usar estadísticas y visualizaciones para descubrir patrones y tendencias.  
5. **Modelado:** aplicar técnicas estadísticas o de *machine learning* según el objetivo.  
6. **Interpretación y comunicación:** explicar resultados de forma clara y visual, con foco en la toma de decisiones.  

> Este flujo no siempre es lineal: los analistas suelen **volver a pasos anteriores** para mejorar la calidad del análisis y la comprensión del problema.

En un DataFrame, cada columna tiene un **tipo de dato** (*dtype*) que determina cómo se almacenan y procesan los valores.  
Los más comunes son:

- **int64:** números enteros (por ejemplo, 1, 25, 100).  
- **float64:** números decimales (por ejemplo, 3.14, 250.5).  
- **object:** texto o datos mixtos (strings).  
- **bool:** valores lógicos `True` o `False`.  
- **datetime64:** fechas y tiempos (`YYYY-MM-DD HH:MM:SS`).  
- **category:** variables con categorías fijas, útiles para optimizar memoria.

> 📊 Puedes consultar los tipos de datos de tu DataFrame con `openaq_pm25.dtypes` o convertirlos con `openaq_pm25.astype()`.

### 📊 Exploración de estadísticas

````python
# Calcular estadísticas básicas include='all'
openaq_pm25.describe()
````

### 🧩 Manejando datos

**Renombrar columnas**

````python
openaq_pm25.rename(columns={"value": "pm25_level"})
````

El parámetro **`inplace=True`** indica que la operación debe aplicarse **directamente sobre el DataFrame original**, sin necesidad de reasignarlo a una nueva variable.  
Por ejemplo, al ejecutar `openaq_pm25.dropna(inplace=True)`, las filas con valores nulos se eliminan del propio `openaq_pm25.`.  
Si en cambio usamos `inplace=False` (valor por defecto), la función devuelve una **copia modificada** del DataFrame, dejando el original sin cambios.  
> 💡 Úsalo con cuidado: `inplace=True` ahorra memoria, pero puede dificultar deshacer cambios si necesitas conservar la versión anterior del DataFrame.

````python
openaq_pm25.rename(columns={"value": "pm25_level"},inplace=True)
````

**Reordenar o seleccionar columnas**

````python
openaq_pm25_selection=openaq_pm25[['country_id', 'pm25_level', 'unit']]
openaq_pm25_selection
````

**Eliminar columnas**


````python
openaq_pm25.drop(columns=["sexo"], inplace=True) 
openaq_pm25.head()
````

### 🧼 Limpieza y transformación de datos

**Manejo de valores nulos/faltantes**

````python
# Eliminar filas con valores nulos
openaq_pm25.dropna(inplace=True)
````

````python
# Reemplazar valores nulos con la media
openaq_pm25["diagnóstico_confirmado"].fillna('descononocido', inplace=True)

````

**Manejo de duplicados**

````python
# Detectar duplicados
openaq_pm25.duplicated().sum()

# Eliminar duplicados
openaq_pm25.drop_duplicates(inplace=True)
````

**Formatear datos**


````python
# Convertir texto a minúsculas
openaq_pm25["sintoma_principal"] = openaq_pm25["sintoma_principal"].str.lower()

# Cambiar tipo de dato
openaq_pm25["edad"] = openaq_pm25["edad"].astype(float)
````

### 🔎 Filtrar, ordenar y resumir datos

**Filtrar datos en pandas**

````python
# Filtrar pacientes con mas de 40 años
openaq_pm25._filtrado = openaq_pm25[openaq_pm25["edad"] > 40]
openaq_pm25._filtrado
````

**Ordenar datos**

````python
# Ordenar de mayor a menor
openaq_pm25.sort_values(by="region", ascending=False)
````

### 🔗 Combinar y estructurar datasets

**Combinar DataFrames (merge)**

````python
clientes = pd.DataFrame({
    "cliente": ["ana", "luis", "maría", "carlos"],
    "ciudad": ["CDMX", "Monterrey", "Guadalajara", "Puebla"]
})

ventas = pd.DataFrame({
    "cliente": ["ana", "luis", "maría", "carlos"],
    "monto_ventas": [250, 400, 150, 300]
})

openaq_pm25._combinado = pd.merge(clientes, ventas, on="cliente")
openaq_pm25._combinado
````

**Ordenar datasets**


````python
# Ordenar alfabéticamente por cliente
openaq_pm25._combinado.sort_values(by="cliente", inplace=True)
openaq_pm25._combinado
````
