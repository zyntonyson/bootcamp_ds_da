## 🧠 Pruebas de hipótesis

En análisis de datos casi nunca tenemos a toda la población. Trabajamos con **muestras**, y eso implica **incertidumbre**.

Si queremos estimar un parámetro poblacional (por ejemplo, la media de ventas), aceptamos que existe un margen de error.

Una **prueba de hipótesis** nos permite cuantificar qué tan compatible es nuestra muestra con una afirmación inicial.

> No probamos que algo sea verdadero. Probamos si tenemos suficiente evidencia para rechazar una afirmación inicial.



### 🧩 Elementos de una prueba de hipótesis

**1️⃣ Muestra aleatoria significativa**  
Debe representar adecuadamente a la población. Tamaño suficiente para poder representar el valor poblacional que estamos estimando.

**2️⃣ Hipótesis claras**  

- $H_0$: hipótesis nula (no hay efecto / no hay diferencia/*status quo*)  
- $H_1$: hipótesis alternativa (sí hay efecto / diferencia)

**3️⃣ Nivel de significancia ($\alpha$)**  
Máximo riesgo aceptado (probabilidad) de aceptar un cambio cuando no existe.  Comúnmente: $\alpha = 0.05$

**4️⃣ p-value**  
Probabilidad de observar un resultado igual o más extremo que el observado, si $H_0$ fuera verdadera.

Regla de decisión:

- Si $p < \alpha$ → Rechazamos $H_0$
- Si $p \geq \alpha$ → No rechazamos $H_0$

**Ejemplo 1: Incremento en ticket promedio después de rediseño web**   

Una empresa de e-commerce rediseña su página principal con el objetivo de aumentar el ticket promedio. El ticket histórico es de 500 pesos. La decisión es relevante porque un incremento sostenido impactaría directamente en los ingresos anuales y en la estrategia de inversión en UX.

**Datos:**  
Disponemos de una muestra de compras realizadas después del rediseño.

- **Hipótesis nula (H₀):** El ticket promedio después del rediseño es igual o menor a 500 pesos ($\mu \leq 500$).  
  En palabras: el rediseño no mejora el ticket promedio.

- **Hipótesis alternativa (H₁):** El ticket promedio después del rediseño es mayor a 500 pesos ($\mu > 500$).  
  En palabras: el rediseño sí aumenta el ticket promedio.

- **Nivel de significancia ($\alpha$):** 0.05  

- **Regla de decisión:**  
  - Si el valor-p < 0.05, rechazamos H₀ (hay evidencia de incremento).  
 

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
np.random.seed(1)
sns.set_style('whitegrid')
sns.set_palette('pastel')
```

```python
muestra = np.random.normal(520, 60, 40)
alpha = 0.05
```

```python
media_muestral = muestra.mean()
std_muestral = muestra.std()
media_esperada = 500
sns.histplot(muestra,kde=True)
plt.axvline(media_muestral, color='red', linestyle='--',
            label=f"Media muestral = {media_muestral:.2f}\nStd muestral = {std_muestral:.2f}")

plt.axvline(media_esperada, color='blue', linestyle='-',
            label=f"Media esperada = {media_esperada:.2f}")
plt.legend()
plt.title('Distribución de los ticket monto del ticket en la muestra')

```

```python
sns.boxplot(x=muestra)
plt.title('Distribución de los ticket monto del ticket en la muestra')

```

> ¿Por que es importante revisar valores atípicos? 🤔

```python
#Prueba unilateral de una muestra
t_stat, p_value = stats.ttest_1samp(muestra, 500)
p_value = p_value / 2

print(f"Tenemos un p-value de {p_value:.2f} entonces dado que", end=' ')
if p_value< alpha:
   print( f"es menor que la significancia ({alpha}). Rechazamos la hipotesis nula, Hay evidencia de que ticket sea 500 o menor")
else:
    print(f"es mayor que la significancia ({alpha}). No Rechazamos la hipotesis nula, no hay evidencia de que el  ticket puede ser de  500 o menos")
```

**Ejemplo 2: Reducción en tiempo de carga de aplicación**

El tiempo histórico promedio de carga es 3.5 segundos. Se implementó una mejora técnica y se desea validar si el tiempo promedio disminuyó.

**Hipótesis**

- **H₀:** El tiempo promedio es mayor o igual a 3.5 segundos ($\mu \geq 3.5$).  
  En palabras: la mejora no redujo el tiempo.

- **H₁:** El tiempo promedio es menor a 3.5 segundos ($\mu < 3.5$).  
  En palabras: la mejora sí redujo el tiempo.

**Significancia:** $\alpha = 0.05$

**Regla de decisión:**  
Si el valor-p es menor que 0.05, rechazamos H₀ y concluimos que el tiempo disminuyó.  
Si el valor-p es mayor o igual que 0.05, no rechazamos H₀.

```python
muestra = np.random.normal(3.2, 0.5, 35)
```

```python
sns.boxplot(x=muestra)
plt.title('Distribución de los tiempos de carga')
```

```python
media_muestral = muestra.mean()
std_muestral = muestra.std()
media_esperada = 3.5
sns.histplot(muestra,kde=True)
plt.axvline(media_muestral, color='red', linestyle='--',
            label=f"Media muestral = {media_muestral:.2f}\nStd muestral = {std_muestral:.2f}")

plt.axvline(media_esperada, color='blue', linestyle='-',
            label=f"Media esperada = {media_esperada:.2f}")
plt.legend()
plt.title('Distribución de los tiempos de carga en la muestra')
```

```python
t_stat, p_value = stats.ttest_1samp(muestra, 3.5)
p_value = p_value / 2  # cola izquierda


print(f"Tenemos un p-value de {p_value:.2f} entonces dado que", end=' ')
if p_value< alpha:
   print( f"es menor que la significancia ({alpha}). Rechazamos la hipotesis nula, no hay evidencia de que el tiempo haya disminuido")
else:
    print(f"es mayor que la significancia ({alpha}). No Rechazamos la hipotesis nula,  hay evidencia de que el tiempo ha disminuido")
```

**Ejemplo 3: Diferencia salarial entre dos áreas (muestras independientes)**

Se comparan los salarios promedio entre el área A y el área B para evaluar si existe diferencia.

**Hipótesis**

- **H₀:** Los salarios promedio son iguales ($\mu_A = \mu_B$).  
  En palabras: no hay diferencia salarial.

- **H₁:** Los salarios promedio son diferentes ($\mu_A \neq \mu_B$).  
  En palabras: sí hay diferencia salarial.

**Significancia:** $\alpha = 0.05$

**Regla de decisión:**  
Si el valor-p es menor que 0.05, rechazamos H₀ y concluimos que existe diferencia significativa.  
Si el valor-p es mayor o igual que 0.05, no rechazamos H₀.

```python
grupo_A = np.random.normal(25000, 4000, 30)
grupo_B = np.random.normal(27000, 4500, 35)
```

```python
sns.histplot(grupo_A, kde=True, stat="density", alpha=0.5, label="Grupo A")
sns.histplot(grupo_B, kde=True, stat="density", alpha=0.5, label="Grupo B")

plt.title("Distribución de salarios")
plt.xlabel("Salario")
plt.legend()
plt.show()
```

```python
df = pd.DataFrame({
    "salario": np.concatenate([grupo_A, grupo_B]),
    "grupo": ["A"]*len(grupo_A) + ["B"]*len(grupo_B)
})

plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="grupo", y="salario")
plt.title("Comparación de salarios por grupo")
plt.ylabel("Salario")
plt.legend()
plt.show()
```

```python
t_stat, p_value = stats.ttest_ind(grupo_A, grupo_B)

print(f"Tenemos un p-value de {p_value:.2f} entonces dado que", end=' ')
if p_value< alpha:
   print( f"es menor que la significancia ({alpha}). Rechazamos la hipotesis nula,  hay evidencia de diferencia en los salarios")
else:
    print(f"es mayor que la significancia ({alpha}). No Rechazamos la hipotesis nula,  no hay evidencia de diferencia en los salarios")
```

**Ejemplo 4: Fertilizante nuevo aumenta producción (una cola derecha, muestras independientes)**

Una empresa agrícola está probando un fertilizante nuevo.  
El fertilizante tradicional produce en promedio 110 kg por parcela.  
Si el nuevo fertilizante aumenta significativamente la producción, la empresa migrará completamente al nuevo proveedor.

Se recolectan datos de producción en parcelas independientes para ambos fertilizantes.

**Hipótesis**

- **Hipótesis nula (H₀):** La producción promedio del fertilizante nuevo es menor o igual que la del tradicional ($\mu_{nuevo} \leq \mu_{tradicional}$).  
  En palabras: el fertilizante nuevo no mejora la producción.

- **Hipótesis alternativa (H₁):** La producción promedio del fertilizante nuevo es mayor que la del tradicional ($\mu_{nuevo} > \mu_{tradicional}$).  
  En palabras: el fertilizante nuevo sí aumenta la producción.

**Significancia:** $\alpha = 0.05$

**Regla de decisión:**  
Si el valor-p unilateral es menor que 0.05 y la media del fertilizante nuevo es mayor que la del tradicional, rechazamos la hipótesis nula y concluimos que el nuevo fertilizante incrementa la producción.  
Si el valor-p es mayor o igual que 0.05, no rechazamos la hipótesis nula.

```python

nuevo = np.random.normal(120, 8, 50)
trad = np.random.normal(110, 15, 50)
```

```python
sns.histplot(nuevo, kde=True, stat="density", alpha=0.5, label="Nuevo")
sns.histplot(trad, kde=True, stat="density", alpha=0.5, label="Tradicional")

plt.title("Distribución de la producción")
plt.xlabel("Kilogramos")
plt.legend()
plt.show()
```

```python
df = pd.DataFrame({
    "producción": np.concatenate([nuevo, trad]),
    "grupo": ["Nuevo"]*len(nuevo) + ["Tradicional"]*len(trad)
})

plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="grupo", y="producción")

plt.title("Distribución de la producción")
plt.xlabel("Kilogramos")
plt.legend()
plt.show()
```

```python
alpha = 0.05


t_stat, p_value = stats.ttest_ind(nuevo, trad,equal_var=False)
p_value = p_value / 2  # cola derecha

print(f"Tenemos un p-value de {p_value:.2f} entonces dado que", end=' ')
if p_value< alpha:
   print( f"es menor que la significancia ({alpha}). Rechazamos la hipotesis nula,  hay evidencia de que el nuevo fertilizante de mejores resultados de producción")
else:
    print(f"es mayor que la significancia ({alpha}). No Rechazamos la hipotesis nula,  no hay evidencia de que el nuevo fertilizante de mejores resultados de producción")
```

**Ejemplo 6: Método reduce errores de digitación (una cola izquierda, muestras independientes)**

Una empresa implementa un nuevo sistema de captura de datos.  
Históricamente, el sistema antiguo produce en promedio 15 errores por jornada.  
Se desea validar si el nuevo sistema reduce el número promedio de errores.

**Hipótesis**

- **H₀:** El número promedio de errores con el nuevo sistema es mayor o igual que con el antiguo ($\mu_{nuevo} \geq \mu_{antiguo}$).  
  En palabras: el nuevo sistema no reduce errores.

- **H₁:** El número promedio de errores con el nuevo sistema es menor que con el antiguo ($\mu_{nuevo} < \mu_{antiguo}$).  
  En palabras: el nuevo sistema sí reduce errores.

**Significancia:** $\alpha = 0.05$

**Regla de decisión:**  
Si el valor-p unilateral es menor que 0.05 y la media del nuevo sistema es menor que la del antiguo, rechazamos H₀ y concluimos que el nuevo método reduce errores.  
Si el valor-p es mayor o igual que 0.05, no rechazamos H₀.

```python
antiguo = np.random.normal(15, 9, 30)
nuevo = np.random.normal(15, 5, 30)
```


#### Prueba de proporciones (Prueba Z)

**Ejemplo 1: Incremento en tasa de conversión después de rediseño**

Una empresa digital tiene una tasa histórica de conversión del 10%.  
Después de un rediseño, se desea evaluar si la nueva tasa es mayor.  
La decisión impacta directamente en inversión publicitaria y escalamiento del producto.

Se observa que, de 500 visitantes, 65 realizaron una compra.

**Hipótesis**

- **Hipótesis nula (H₀):** La proporción de conversión es menor o igual a 0.10 ($p \leq 0.10$).  
  En palabras: el rediseño no mejora la tasa de conversión.

- **Hipótesis alternativa (H₁):** La proporción de conversión es mayor a 0.10 ($p > 0.10$).  
  En palabras: el rediseño sí mejora la tasa de conversión.

**Significancia:** $\alpha = 0.05$

**Regla de decisión:**  
Si el valor-p unilateral es menor que 0.05, rechazamos H₀ y concluimos que la tasa de conversión aumentó significativamente.  
Si el valor-p es mayor o igual que 0.05, no rechazamos H₀.

```python
from statsmodels.stats.proportion import proportions_ztest
```

```python
alpha = 0.05

exitos = 65
n = 500
p0 = 0.10

z_stat, p_value = proportions_ztest(count=exitos, nobs=n, value=p0, alternative='larger')




print(f"Tenemos un p-value de {p_value:.2f} entonces dado que", end=' ')
if p_value< alpha:
   print( f"es menor que la significancia ({alpha}). Rechazamos la hipotesis nula,  hay evidencia de que la tasa de conversión aumentó")
else:
    print(f"es mayor que la significancia ({alpha}). No Rechazamos la hipotesis nula,  no hay evidencia de que la tasa de conversión aumentó")
```


**Ejemplo 2: Reducción en tasa de defectos en producción**

Una fábrica tiene una tasa histórica de defectos del 8%.  
Tras implementar un nuevo sistema de control de calidad, se desea evaluar si la proporción de defectos disminuyó.

En una muestra de 400 productos, se observan 20 defectuosos.

**Hipótesis**

- **H₀:** La proporción de defectos es mayor o igual a 0.08 ($p \geq 0.08$).  
  En palabras: el nuevo sistema no reduce defectos.

- **H₁:** La proporción de defectos es menor a 0.08 ($p < 0.08$).  
  En palabras: el nuevo sistema sí reduce defectos.

```python
alpha = 0.05

exitos = 20   # defectuosos
n = 400
p0 = 0.08

z_stat, p_value = proportions_ztest(count=exitos, nobs=n, value=p0, alternative='smaller')

print(f"Tenemos un p-value de {p_value:.2f} entonces dado que", end=' ')
if p_value< alpha:
   print( f"es menor que la significancia ({alpha}). Rechazamos la hipotesis nula,  hay evidencia de que la tasa de defectos disminuyó")
else:
    print(f"es mayor que la significancia ({alpha}). No Rechazamos la hipotesis nula,  no hay evidencia de que la tasa de defectos disminuyó")
```

**Ejemplo 3: Comparación de proporciones entre dos campañas (dos colas)**

Una empresa lanza dos campañas digitales (A y B).  
Se desea evaluar si existe diferencia significativa en la proporción de usuarios que realizan una compra.

Campaña A: 120 conversiones de 1000 usuarios.  
Campaña B: 150 conversiones de 950 usuarios.

La decisión impactará en la asignación futura de presupuesto.

**Hipótesis**

- **H₀:** Las proporciones de conversión son iguales ($p_A = p_B$).  
  En palabras: ambas campañas tienen el mismo desempeño.

- **H₁:** Las proporciones de conversión son diferentes ($p_A \neq p_B$).  
  En palabras: existe diferencia entre campañas.

**Significancia:** $\alpha = 0.05$

**Regla de decisión:**  
Si el valor-p es menor que 0.05, rechazamos H₀ y concluimos que existe una diferencia significativa entre las campañas.  
Si el valor-p es mayor o igual que 0.05, no rechazamos H₀.

```python
# Crear datos dicotómicos (1 = éxito, 0 = fracaso)

camp_A = pd.DataFrame({
    "campaña": "A",
    "conversion": np.concatenate([np.ones(120), np.zeros(880)])
})

camp_B = pd.DataFrame({
    "campaña": "B",
    "conversion": np.concatenate([np.ones(150), np.zeros(800)])
})

# Unir en un solo DataFrame
df = pd.concat([camp_A, camp_B], ignore_index=True)

df.head()
```

```python
alpha = 0.05
resumen = df.groupby("campaña")["conversion"].agg(["sum", "count"])
resumen
```

```python
exitos = resumen["sum"].values
n = resumen["count"].values

z_stat, p_value = proportions_ztest(count=exitos, nobs=n, alternative='two-sided')

print(f"Tenemos un p-value de {p_value:.2f} entonces dado que", end=' ')
if p_value< alpha:
   print( f"es menor que la significancia ({alpha}). Rechazamos la hipotesis nula,  hay evidencia de que existe diferencia significativa entre campañas")
else:
    print(f"es mayor que la significancia ({alpha}). No Rechazamos la hipotesis nula,  no hay evidencia de que existe diferencia significativa entre campañas")
```

#### Prueba Chi-Cuadrada de Independencia

 **Ejemplo 1: Canal de adquisición vs Compra (conteos directos)**

Una empresa quiere saber si el canal de adquisición (Orgánico vs Pagado) está relacionado con la decisión de compra.

Se observa la siguiente tabla:

| Canal      | Compra | No compra |
|------------|--------|-----------|
| Orgánico   | 120    | 80        |
| Pagado     | 150    | 50        |

La decisión estratégica depende de saber si el canal influye en la conversión.



**Hipótesis**

- **H₀:** El canal de adquisición y la compra son independientes.  
  En palabras: el canal no influye en la decisión de compra.

- **H₁:** El canal de adquisición y la compra no son independientes.  
  En palabras: sí existe relación entre canal y compra.

**Significancia:** $\alpha = 0.05$

**Regla de decisión:**  
Si el valor-p es menor que 0.05, rechazamos H₀ y concluimos que existe relación significativa.  
Si el valor-p es mayor o igual que 0.05, no rechazamos H₀.

```python
from scipy.stats import chi2_contingency
```

```python
tabla = np.array([[120, 80],
                  [150, 50]])

chi2, p_value, dof, expected = chi2_contingency(tabla)

alpha = 0.05


print(f"Tenemos un p-value de {p_value:.2f} entonces dado que", end=' ')
if p_value< alpha:
   print( f"es menor que la significancia ({alpha}). Rechazamos la hipotesis nula,  hay evidencia de que existe relación entre canal y compra")
else:
    print(f"es mayor que la significancia ({alpha}). No Rechazamos la hipotesis nula,  no hay evidencia de que existe relación entre canal y compra")
```
