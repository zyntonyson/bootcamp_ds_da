# Introducción


Una API, o **Interfaz de Programación de Aplicaciones**, facilita la comunicación entre dos piezas de software. Te permite recuperar y enviar datos utilizando código. Mayormente usamos las API para recuperar datos, por lo cual nos concentraremos en esa parte de las API's


Cuando queremos recibir datos de una API, necesitamos hacer una solicitud. Las solicitudes se utilizan en todo el web. Por ejemplo, cuando visitas la pagina del bootcamp, tu navegador realiza una solicitud solicitud al servidor web de tripleten, el cual responde con los contenidos.


Las solicitudes a las API funcionan exactamente de la misma manera: haces una solicitud a un servidor de API para obtener datos, y responde a tu solicitud.

## Diferentes métodos HTTP y codigos de estatus


hay varios métodos HTTP para las API REST. Estos métodos indican a la API qué operaciones deben realizarse en los datos. Si bien hay muchos métodos HTTP, los cinco métodos enumerados a continuación son los más comúnmente utilizados con las API REST:

- **GET**: Este método se utiliza para solicitar datos de un recurso específico. Por ejemplo, al realizar una solicitud GET a la URL de un recurso, la API devolverá los datos de ese recurso.

- **POST**: Se utiliza para enviar datos a un recurso específico para su procesamiento o almacenamiento. Por ejemplo, al realizar una solicitud POST a la URL de un recurso, la API creará un nuevo recurso utilizando los datos proporcionados en el cuerpo de la solicitud.

- **PUT**: Se utiliza para actualizar los datos de un recurso específico. Al realizar una solicitud PUT a la URL de un recurso, la API actualizará los datos de ese recurso con los datos proporcionados en el cuerpo de la solicitud.

- **DELETE** : Se utiliza para eliminar un recurso específico. Al realizar una solicitud DELETE a la URL de un recurso, la API eliminará ese recurso.

- **PATCH**: Se utiliza para actualizar parcialmente los datos de un recurso específico. A diferencia del método PUT, que actualiza todos los datos de un recurso, el método PATCH permite actualizar solo parte de los datos de un recurso.

Una vez que una API REST recibe y procesa una solicitud HTTP, devuelve una respuesta con un código de estado HTTP. Este código de estado proporciona información sobre la respuesta y ayuda a la aplicación cliente a saber qué tipo de respuesta es.

Los códigos de estado están numerados en función de la categoría del resultado:

- 1xx: Información - Indica que la solicitud ha sido recibida y el servidor continúa procesando la solicitud.

- 2xx: Éxito - Indica que la acción solicitada se ha completado con éxito.

- 3xx: Redirección - Indica que se requiere una acción adicional para completar la solicitud.

- 4xx: Error del cliente - Indica que hubo un error por parte del cliente al realizar la solicitud.

- 5xx: Error del servidor - Indica que hubo un error en el servidor al intentar completar la solicitud del cliente.

# Tutorial: Realizar consultas a un endpoint de una API


Los puntos finales de API son las URL públicas expuestas por el servidor que una aplicación cliente utiliza para acceder a recursos y datos.

Para este tutorial, estaremos utilizando la [API REST de Fake Store](https://fakestoreapi.com/). Más específicamente, estaremos utilizando los siguientes puntos finales:

- **GET** 
    - /products: Obtiene la lista de productos.
    - /products?limit=x: Obtiene solo 5 productos.
    - /products/<product_id>:obtiene un solo product.
- **POST**
    - /products: Crea un nuevo producto.
- **PUT**
    - /products/<product_id>: actualiza un producto.
- **PATCH**
    - /products/<product_id>: actualiza parcialmente un producto.
- **DELETE**
    - /products/<product_id>: elimina un producto


Los elementos mostrados debajo de cada comando son los **ENDPOINTS** de la API, los que realizan diferentes acciones basados en el método HTTP asociado. 

Para realizar consultas, necesitaremos la URL base de la api, la cual es: https://fakestoreapi.com

## Realizar un request con GET

```python
import requests

BASE_URL = 'https://fakestoreapi.com'

response = requests.get(f"{BASE_URL}/products")
display(response.json())
```


El script anterior utiliza el método requests.get() para enviar una solicitud GET al punto final de la API /products. Responde con una lista de todos los productos. Luego llamamos a .json() para ver la respuesta JSON.

formato de intercambio de datos muy popular para las API REST.

También puedes imprimir otros atributos relacionados con la respuesta, como el código de estado.






```python
print(response.status_code)
```

Como sabemos, el código de estado 200 significa una respuesta exitosa.

Dado que el punto final /products devuelve muchos datos, limitemos estos datos a solo 3 productos.

Para hacer esto, tenemos un punto final /products?limit=x donde x es un número entero positivo. El límite se llama parámetro de consulta. Veamos cómo podemos agregar este parámetro de consulta en la solicitud.






```python
query_params = {
    "limit": 3
}

response = requests.get(f"{BASE_URL}/products?limit=3")
display(response.json())
```

```python
productos = response.json()
productos[0]
```

```python
query_params = {
    "limit": 3
}

response = requests.get(f"{BASE_URL}/products", params=query_params)
display(response.json())
```


El método requests.get() toma un parámetro llamado params donde podemos especificar nuestros parámetros de consulta en forma de un diccionario de Python. Por lo tanto, hemos creado un diccionario llamado query_params y pasamos limit como la clave y 3 como el valor. Luego pasamos estos query_params en requests.get().

Ahora tenemos los datos de respuesta limitados a solo 3 productos. Intentemos obtener solo un producto con el id 18.

```python
response = requests.get(f"{BASE_URL}/products/2")
display(response.json())
```

Dado que tenemos un punto final /products/<product_id>, podemos pasar el id 18 en la URL de la API y hacer una solicitud GET en él.

## Realizar un Request con POST

Usamos la solicitud POST para agregar nuevos datos a la API REST. Los datos se envían al servidor en formato JSON que se parece a un diccionario de Python. Según la documentación de la API de Fake Store, un producto tiene los siguientes atributos: título, precio, descripción, imagen y categoría. Entonces, un nuevo producto se vería así:

```python
new_product = {
    "title": 'test product',
    "price": 13.5,
    "description": 'lorem ipsum set',
    "image": 'https://i.pravatar.cc',
    "category": 'electronic'
}
```

Podemos enviar una solicitud POST para añadir este nuevo producto usando el método request.post() de la siguiente forma:

```python
response = requests.post(f"{BASE_URL}/products", json=new_product)
display(response.json())
```


En el método requests.post(), podemos pasar datos JSON usando el argumento json. El uso del argumento json establece automáticamente el Content-Type en Application/JSON en el encabezado de la solicitud.

Una vez que hacemos una solicitud POST en el punto final /products, obtenemos un objeto de producto con el id en la respuesta.

## Realizar un request con PUT

A menudo necesitamos actualizar datos existentes en la API. Usando la solicitud PUT, podemos actualizar los datos completos. Esto significa que cuando hacemos una solicitud PUT, reemplaza los datos antiguos con los nuevos.

En la solicitud POST, creamos un nuevo producto cuyo id era 21. Actualicemos el antiguo producto con un nuevo producto haciendo una solicitud PUT en el punto final products/<product_id>.

```python
updated_product = {
    "title": 'updated_product',
    "category": 'clothing'
}

response = requests.put(f"{BASE_URL}/products/21", json=updated_product)
display(response.json())
```

##

## Realizando un request con PATCH

A veces, no necesitamos reemplazar completamente los datos antiguos. Más bien, deseamos modificar solo ciertos campos. En ese caso, usamos la solicitud PATCH.

Actualicemos la categoría del producto de nuevo de ropa a electrónica haciendo una solicitud PATCH en el punto final products/<product_id>.

```python
updated_product = {
    "category": 'videogames'
}

response = requests.patch(f"{BASE_URL}/products/21", json=updated_product)
display(response.json())
```

## Realizando un request con DELETE


Como sugiere el nombre, si deseas eliminar un recurso de la API, puedes usar una solicitud DELETE. Eliminemos este producto con el id 21.

```python


response = requests.delete(f"{BASE_URL}/products/21")
print(response.json())
```
