# 📘 Cuestionario de repaso

## Pregunta 1 - 💻 Comandos en terminal

❓ ¿Cuál es el comando en la terminal para listar los archivos y carpetas dentro de un directorio?

- 📂 ls  
- 🔄 cd  
- 📁 mkdir  
- 📄 touch  

### ✅ Respuesta correcta
La respuesta es **ls**, porque este comando muestra el contenido del directorio actual, mientras que *cd* cambia de directorio, *mkdir* crea carpetas y *touch* crea archivos.

## Pregunta 2 - 🌿 Git y control de versiones

❓ ¿Cuál de las siguientes acciones describe mejor el comando `git commit`?

- ⬇️ Descarga los cambios desde GitHub hacia tu repositorio local  
- ⬆️ Envía los cambios locales al repositorio remoto  
- 💾 Guarda una instantánea de los cambios en el repositorio local  
- 🌱 Crea una nueva rama en el proyecto  

### ✅ Respuesta correcta
La respuesta es **Guarda una instantánea de los cambios en el repositorio local**, porque `git commit` registra los cambios preparados en el historial local antes de subirlos a un remoto.

## Pregunta 3 - 🐍 Python y librerías

❓ En Python, ¿qué utilidad tiene la librería `argparse`?

- 🗄️ Permite ejecutar consultas SQL desde Python  
- ⚙️ Facilita el manejo de argumentos desde la terminal  
- 🖼️ Sirve para crear interfaces gráficas simples  
- 🛠️ Genera entornos virtuales automáticamente  

### ✅ Respuesta correcta
La respuesta es **Facilita el manejo de argumentos desde la terminal**, porque `argparse` permite que un script reciba parámetros y opciones al ejecutarse desde línea de comandos.

## Pregunta 4 - 🏗️ Clases en Python

❓ ¿Cuál es la forma correcta de definir una clase en Python?

-  `class Persona: pass`  
-  `def Persona(): pass`  
-  `new Persona {}`  
-  `create class Persona()`  

### ✅ Respuesta correcta


La respuesta es **class Persona: pass**, porque en Python se definen clases con la palabra clave `class`. Las demás opciones no corresponden a la sintaxis válida en Python.
