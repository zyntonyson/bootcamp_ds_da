## 🎯 Convertir necesidades del negocio en preguntas de datos

> “No empieces con los datos, empieza con la decisión.”

Cuando un área del negocio plantea un problema, nuestro trabajo analítico comienza **traduciendo esa necesidad en una pregunta medible**.

**1️⃣ Pasos clave para enmarcar una pregunta analítica:**
1. **Identifica al stakeholder:** ¿Quién tomará la decisión con la información?  
   *Ejemplo:* Gerente de Ventas.
2. **Define la decisión que se tomará:** ¿Qué acción depende del análisis?  
   *Ejemplo:* Reasignar presupuesto regional.
3. **Delimita el horizonte temporal:** ¿Sobre qué periodo se decidirá?  
   *Ejemplo:* Trimestre actual vs. trimestre anterior.
4. **Explora los datos disponibles:** ¿Qué variables pueden responder la pregunta?  
   *Ejemplo:* `Date, Region, Total_Sales`.

**💡 Ejemplo completo**
> Stakeholder: Gerente de Ventas  
> Necesidad: Entender qué región tiene menor retorno por inversión.  
> Pregunta analítica: *¿Cuál es la rentabilidad promedio por región en el último trimestre?*

## 📊 Tipos de Métricas — Absolutas, Relativas y KPI/OKR

No todas las métricas dicen lo mismo. Diferenciar sus tipos evita conclusiones engañosas.

**1️⃣ Métricas absolutas**
- Son conteos o totales sin referencia a otro valor.  
  *Ejemplo:* “Vendimos **5000 unidades** en mayo.”

**2️⃣ Métricas relativas**
- Comparan una magnitud contra otra o contra el tiempo.  
  *Ejemplo:* “Las ventas crecieron **+10%** respecto a abril.”

**3️⃣ KPI (Key Performance Indicators)**
- Métricas **clave** que miden el desempeño frente a un objetivo.  
  *Ejemplo:* “Tasa de conversión de clientes = 15% (objetivo: 20%).”

**4️⃣ OKR (Objectives and Key Results)**
- Objetivo → cualitativo.  
- Resultados clave → cuantitativos.  
  *Ejemplo:*  
  - **Objetivo:** Mejorar la retención de usuarios.  
  - **KR1:** Aumentar retención mensual del 60% al 75%.  
  - **KR2:** Reducir cancelaciones por inactividad en 20%.

**5️⃣ Guardrail Metrics**
- Métricas de seguridad que aseguran que las mejoras no afecten otros aspectos.  
  *Ejemplo:* Aumentar retención sin afectar satisfacción del cliente (NPS).

## ⏳ Leading vs Lagging Indicators — Mirando causa y efecto

No todas las métricas se comportan igual en el tiempo.  
Algunas *predicen* tendencias, otras *confirman* resultados.

**1️⃣ Leading Indicators (indicadores adelantados)**
- Señalan posibles cambios futuros.  
- Son útiles para tomar acción temprana.  
  *Ejemplo:* “Número de visitas a la app” puede anticipar “nuevos suscriptores”.

**2️⃣ Lagging Indicators (indicadores rezagados)**
- Reflejan el resultado final de acciones pasadas.  
  *Ejemplo:* “Ingresos mensuales” o “tasa de rotación” son efectos ya consolidados.





**💡 Ejemplo combinado**
> En una campaña de retención:  
> - *Leading:* número de usuarios que abren el correo de retención.  
> - *Lagging:* porcentaje de usuarios que renuevan su suscripción.

Hacer preguntas vagas como *“¿Cómo vamos en ventas?”* puede parecer un buen punto de partida, pero en análisis de datos **no conducen a respuestas accionables**. Una buena pregunta analítica debe ser *específica, medible y contextual*. Es decir, debe indicar **qué métrica** se quiere analizar, **para qué grupo o segmento**, y **en qué periodo o contexto temporal**.  
Por ejemplo, en lugar de preguntar “¿Qué región es la mejor?”, podríamos decir “¿Cuál fue el promedio de ventas por región durante el segundo trimestre de 2024?”.  
Esa precisión permite identificar la fuente de datos, elegir las funciones correctas (como un *pivot table* o *groupby*) y obtener información que realmente ayude a tomar decisiones. En síntesis, **una buena pregunta guía el análisis; una vaga solo genera confusión.** 🎯

## 🧠 Glosario de términos clave en análisis de negocio

A continuación encontrarás una lista de conceptos que suelen aparecer con frecuencia en el análisis de datos aplicado a contextos empresariales.  
La idea es tener un *mini diccionario* para que todos hablemos el mismo idioma. 📘

1. **Stakeholder:** Persona o grupo interesado o afectado por una decisión o proyecto.  
   *Ejemplo:* gerente, cliente, equipo de marketing.

2. **Métrica:** Valor numérico que mide un aspecto específico del negocio o proceso.  
   *Ejemplo:* ventas totales, tasa de conversión, churn rate.

3. **Indicador:** Métrica usada para evaluar el desempeño frente a un objetivo.  
   *Ejemplo:* KPI de retención mensual.

4. **KPI (Key Performance Indicator):** Indicador clave que refleja el progreso hacia una meta crítica del negocio.

5. **OKR (Objectives and Key Results):** Marco de objetivos cualitativos acompañados de resultados cuantitativos medibles.

6. **Guardrail Metric:** Métrica de control que asegura que una mejora no cause daño en otro aspecto.  
   *Ejemplo:* aumentar ingresos sin disminuir satisfacción del cliente.

7. **Insight:** Conclusión significativa obtenida del análisis de datos que aporta valor a la toma de decisiones.

8. **Benchmark:** Punto de referencia usado para comparar resultados.  
   *Ejemplo:* promedio del sector, resultados del trimestre anterior.

9. **Lagging Indicator:** Métrica que muestra el resultado de acciones pasadas.  
   *Ejemplo:* ingresos del mes.

10. **Leading Indicator:** Métrica que anticipa tendencias o resultados futuros.  
   *Ejemplo:* número de nuevos leads generados.

11. **ROI (Return on Investment):** Medida del rendimiento de una inversión.  
   *Fórmula:* $ROI = \frac{\text{ganancia neta}}{\text{inversión total}}$

12. **Segmentación:** División de datos en grupos con características comunes para análisis más precisos.  
   *Ejemplo:* segmentar por región, edad o tipo de producto.

13. **Cohorte:** Grupo de usuarios que comparten una característica en un periodo específico.  
   *Ejemplo:* usuarios que se registraron en enero.

14. **Funnel (Embudo):** Representación del recorrido de un usuario desde el inicio hasta la conversión final.  
   *Ejemplo:* visita → registro → compra.

15. **Metric Tree (Árbol de métricas):** Estructura que conecta una métrica principal con sus componentes o submétricas.  
   *Ejemplo:* ventas totales = precio promedio × número de transacciones.

15. **Metric Tree (Árbol de métricas):** Estructura que conecta una métrica principal con sus componentes o submétricas.  
   *Ejemplo:* ventas totales = precio promedio × número de transacciones.
