# Transformar datos en insights de negocio

---

Durante esta sesión cada equipo representará a un grupo de **analistas de datos** asignado a un cliente distinto.  Cada cliente presenta una situación de negocio realista y un conjunto de datos para resolverla.  El objetivo es que los equipos vivan el proceso completo de *data analytics* de principio a fin. 🚀


## 🧩 *Objetivo general de la dinámica*
Aprender a **interpretar las necesidades de un cliente**, trabajar colaborativamente con los datos, y presentar conclusiones útiles que apoyen la toma de decisiones.  

Cada equipo deberá:

1️⃣ Entender el **contexto y objetivos** de su cliente.  
2️⃣ Formular **preguntas analíticas** alineadas al problema.  
3️⃣ Realizar **limpieza, exploración y análisis** de los datos.  
4️⃣ Elaborar **visualizaciones y métricas clave**.  
5️⃣ Presentar **insights y recomendaciones** al final de la sesión.

### 🩰 Academia de baile “Ritmo Vivo” - Planeación de próximos cursos

---




La **Academia Ritmo Vivo** ofrece clases presenciales y virtuales de baile en distintas disciplinas: salsa, bachata, tango, hip-hop y danza contemporánea.  
Durante el último año, la dirección ha notado **fluctuaciones en la asistencia y preferencia por horarios**, además de un **descenso en la inscripción de nuevos alumnos**.  

El **stakeholder principal** es *Mariana*, la **Directora Académica**, quien debe planificar el nuevo calendario de cursos para el siguiente trimestre.  
Mariana necesita basar sus decisiones en evidencia y datos del ciclo anterior (enero–junio 2024) para optimizar la oferta.  



#### 🎯 Objetivo del análisis  
Identificar qué cursos y horarios presentan mayor demanda y cuáles podrían ajustarse o reemplazarse en el nuevo trimestre.  



#### 🧩 Decisiones que deben tomarse  
1. ¿Qué estilos de baile deben **mantenerse o ampliarse** según el interés de los alumnos?  
2. ¿Qué horarios muestran **mayor asistencia o cancelación** (vespertino/nocturno entre semana/fines de semana)?  
3. ¿Existe diferencia en la preferencia entre **modalidad presencial y virtual**?  
4. ¿Qué instructores tienen mejor tasa de retención de alumnos?  
5. ¿Qué nuevos cursos podría probar la academia basándose en las tendencias actuales?  




#### 💬 Preguntas / necesidades del stakeholder
1. “¿Qué estilos de baile tuvieron mayor ocupación promedio por sesión?”  
2. “¿En qué horarios se concentran más inscripciones y asistencias?”  
3. “¿La modalidad virtual sigue siendo rentable comparada con la presencial?”  
4. “¿Qué instructores tienen más cancelaciones o faltas?”  
5. “¿Podríamos identificar un horario óptimo para lanzar un nuevo curso de salsa o bachata?”  
6. ¿La carga de trabajo de los instructores es adecuada?

### 🌶️ Condimentos  “Sabor Sureste” — Evaluación de expansión de sucursales

---


“Sabor Sureste” es una marca regional de condimentos (mezclas en polvo, salsas y adobos) con operaciones en Yucatán. La dirección busca decidir **en qué ciudades vecinas abrir 1–3 nuevas sucursales** durante el próximo trimestre, priorizando puntos con **alta demanda potencial**, **competencia manejable** y **costos operativos sostenibles**.

#### Stakeholder principal:  
*Laura Hernández — Directora de Expansión.*

#### Objetivo del análisis:  
Estimar la demanda potencial y el tiempo de recuperación por ciudad (break-even), priorizando dónde abrir sucursales físicas y/o fortalecer venta online.

#### Decisiones a tomar:  
1) Top 2–3 ciudades para apertura.  
2) Formato por ciudad: *Tienda física* vs *Punto pickup + refuerzo online*.  
3) Mix de productos (polvos vs salsas) y rango de precios.  
4) Horarios y dotación mínima de personal.  

#### Horizonte temporal:  
Plan de apertura Q1–Q2 2026, con base en datos recopilados en jul–sep 2025.


#### 💬 Preguntas / necesidades del stakeholder
1. ¿Dónde hay **mayor intención de compra** y **frecuencia** declarada?  
2. ¿Cómo varía la **sensibilidad al precio** por ciudad?  
3. ¿Qué ciudades presentan **competencia intensa** (market share y número de tiendas)?  
4. ¿Qué **canales preferidos** (tienda, marketplace, e-commerce propio) dominan por ciudad?  
5. ¿Cuál es el **ingreso estimado por tienda** y el **tiempo a break-even** en los escenarios base/optimista?

### 📺 Empresa de medios “Conecta Media” — Alianzas estratégicas con creadores

---

Conecta Media es una empresa de medios digitales que gestiona campañas para marcas en LATAM.  
Para el próximo trimestre desea **priorizar alianzas con creadores de contenido** que maximicen alcance y retorno, asegurando además la **idoneidad para todo público** cuando aplique. Se analizarán datos de:  
1) **Creadores** (tipo de contenido, aptitud para todo público, audiencia y tarifas),  
2) **Marcas/Clientes** (industria, producto, público objetivo e importancia estratégica), y  
3) **Campañas del último trimestre** (jul–sep 2025: costo, impresiones, ingresos, plataforma y creador).

#### Stakeholder principal:
*Sara López — Directora de Desarrollo de Alianzas.*

#### Objetivos del análisis:  
- Identificar **top creadores** por *fit* con marca, alcance efectivo y costo por resultado.  
- Evaluar **mix de plataformas** (YT/IG/TikTok) y su desempeño por industria.  
- Sugerir **paquetes de alianza** (multi-plataforma/multi-marca) con estimados de retorno.

#### Preguntas/ necesidades del stakeholder: 
1. ¿Qué creadores son más adecuados para **marcas family-safe** y público general?  
2. ¿Qué **costo promedio por campaña** y **CPM/ROI** podemos esperar por plataforma?  
3. ¿Cómo se relaciona el **perfil de audiencia** del creador con el **target** de las marcas?  
4. ¿Qué **marcas Tier A** requieren prioridad y con qué creadores conviene amarrar Q4?  
5. En el trimestre pasado (jul–sep 2025), ¿qué combinaciones **marca-creador-plataforma** generaron mejor **ingreso neto**?

## ⏱️ *Estructura del trabajo*




Cada proyecto seguirá estas etapas guiadas:

1️⃣ **Organización del equipo (5 min)**  
   - Nombrar roles: responsable, analista visual, encargado de bitácora, expositor.  
   - Asignar el cliente y conjunto de datos.  

2️⃣ **Análisis del contexto y revisión de datos (15 min)**  
   - Leer el caso del cliente.  
   - Revisar los datasets entregados.  
   - Formular preguntas al “stakeholder” para aclarar objetivos.  

3️⃣ **Limpieza, transformación y enriquecimiento (15 min)**  
   - Corregir valores inconsistentes, unificar formatos y realizar joins.  
   - Documentar los cambios en la bitácora.  

4️⃣ **Análisis y descubrimiento de insights (15 min)**  
   - Aplicar tablas dinámicas, gráficos y cálculos de métricas.  
   - Identificar patrones, comparaciones y hallazgos.  

5️⃣ **Preparación de presentación (20 min)**  
   - Crear una mini presentación (3–5 diapositivas).  
   - Incluir gráficos, hallazgos y recomendaciones para el cliente.  
   - [Template-presentación](https://docs.google.com/presentation/d/1sf1B7ye5qML7Z3SfHle_mXc3uyveqvg1TQDQ-iAyl9Y/copy)


### 🎤 *Cierre de la sesión*

Cada equipo presentará brevemente sus resultados frente al grupo, simulando una reunión de consultoría.  
Los demás equipos podrán hacer *preguntas o comentarios críticos*, reforzando el aprendizaje colectivo.  

💬 *Recuerda:* no hay una única respuesta correcta. Lo más importante es **el proceso analítico**, la claridad de los argumentos y la capacidad de conectar los datos con decisiones reales del negocio.
