## **Storytelling con dashboards** 📖📊


---


Un dashboard no es solo un conjunto de gráficos: es una herramienta para *contar historias con datos*.  
Más allá de mostrar cifras y tendencias, un buen dashboard guía al espectador a través de un relato claro, con un inicio, desarrollo y conclusión.

Contar una historia con datos significa:
- Conectar la información con un contexto relevante.  
- Responder preguntas clave de manera visual e intuitiva.  
- Guiar la mirada del usuario hacia los hallazgos más importantes.  

Los dashboards facilitan este proceso porque:
- Permiten **interactividad**, lo que da al usuario control sobre qué explorar.  
- Integran múltiples visualizaciones en un solo lugar, ayudando a ver el panorama completo.  
- Ofrecen filtros y segmentaciones para descubrir matices y patrones ocultos.  



### **Recomendaciones para diseñar un dashboard** 🛠️🎨

----

Un dashboard efectivo no se trata de cuántos gráficos tiene, sino de **qué tan bien comunica** la información.  
Al diseñar, ten en cuenta:

- **Claridad: menos es más** ➡️ Evita la saturación visual, prioriza lo esencial.  
- **Consistencia visual** ➡️ Mantén la misma paleta de colores, tipografía y jerarquía visual.  
- **Focaliza en métricas clave (KPIs)** ➡️ Destaca solo los indicadores que realmente importan para el objetivo.  
- **Interactividad útil** ➡️ Usa filtros, drill-downs y botones de navegación que aporten valor al análisis.  


### **Cuándo usar un dashboard** ⏱️📊


---

Un dashboard es ideal cuando se necesita:

- **Monitoreo de KPIs en tiempo real** ➡️ Seguimiento continuo de métricas clave del negocio.  
- **Comparación entre categorías o periodos** ➡️ Evaluar cambios y detectar patrones.  
- **Exploración interactiva** ➡️ Facilitar que el usuario encuentre respuestas a preguntas de negocio de forma dinámica.

*Recuerda:* No todos los reportes necesitan ser dashboards. Si el objetivo es presentar un análisis puntual y estático, un informe escrito o una visualización única puede ser más efectivo.

## **Práctica con PowerBI,Tableau y Looker** 🎮📊

---


Antes de abrir las herramientas, es importante tener claro el **contexto** y la **historia** que queremos contar.  
En nuestro caso, partimos de un conjunto de datos [`games.csv`](https://practicum-content.s3.us-west-1.amazonaws.com/datasets/games.csv) que contiene información sobre videojuegos: año de lanzamiento, plataforma, género, ventas por región y ventas globales.  

**Historia que vamos a contar:**  


*"La industria de los videojuegos ha evolucionado drásticamente en las últimas décadas. ¿Qué géneros han dominado las ventas globales? ¿Cuáles han sido los años más exitosos? ¿Existen diferencias marcadas entre las preferencias de cada región?"*



**Preguntas que nuestro dashboard debe responder:**
- ¿Cómo han cambiado las ventas globales a lo largo del tiempo?  
- ¿Qué géneros han generado mayores ventas históricamente?  
- ¿Qué plataformas son más populares en distintas regiones?  
- ¿Cuál ha sido el mejor año en ventas y cuántos títulos se publicaron?  
- ¿Existen patrones regionales que puedan orientar estrategias de marketing?  



**Métricas deseadas:**  

---


1. ➡️ Ventas totales, número de títulos publicados, mejor año de ventas.
1. ➡️ Top 10 géneros por ventas totales.
1. ➡️ Evolución de ventas globales por año de lanzamiento.  
1. ➡️ Evolución de ventas globales por año de lanzamiento segmentado por plataforma.  
1. **Filtros** ➡️ Año, plataforma y género.
