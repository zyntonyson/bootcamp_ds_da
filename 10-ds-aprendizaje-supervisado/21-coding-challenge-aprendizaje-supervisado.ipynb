## 🧩 Code challenge: **Churn prediction**

> *BetaBank* ha notado una pérdida progresiva de clientes en los últimos meses. El equipo de analítica debe desarrollar un modelo de clasificación que prediga qué clientes tienen mayor probabilidad de cancelar su contrato, permitiendo así acciones preventivas. Tú eres parte del equipo de ciencia de datos responsable de entregar un modelo funcional y bien evaluado.

---

### 📦 Dataset disponible: `churn_dataset`

Contiene información histórica de 10,000 clientes. Cada fila representa un cliente y sus características al momento de la observación.

---

### 📘 Diccionario de datos

| Variable            | Tipo         | Descripción |
|---------------------|--------------|-------------|
| `RowNumber`         | int          | Índice de fila (irrelevante) |
| `CustomerId`        | int          | ID único de cliente |
| `Surname`           | object       | Apellido del cliente |
| `CreditScore`       | int          | Puntaje crediticio |
| `Geography`         | object       | País de residencia |
| `Gender`            | object       | Género del cliente |
| `Age`               | int          | Edad |
| `Tenure`            | float        | Años como cliente |
| `Balance`           | float        | Saldo promedio |
| `NumOfProducts`     | int          | Productos contratados |
| `HasCrCard`         | int (0/1)    | Tiene tarjeta de crédito |
| `IsActiveMember`    | int (0/1)    | Es cliente activo |
| `EstimatedSalary`   | float        | Salario estimado |
| `Exited`            | int (0/1)    | Variable objetivo: 1 = canceló, 0 = se quedó |
